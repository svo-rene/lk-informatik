## Was ist ein Stack?

Ein Stack ist eine dynamische Datenstruktur, die dem Prinzip **Last In, First Out** (LIFO) folgt. 
Das bedeutet, dass das zuletzt hinzugefügte Element als erstes wieder entfernt wird.

Einige reale Beispiele für Stacks:
- Ein Stapel Bücher, bei dem das oberste Buch als erstes entfernt wird.
- Ein Web-Browser-Verlauf, bei dem man zur zuletzt besuchten Seite zurückkehren kann.

### Grundlegende Operationen auf einem Stack
1. **Push**: Ein neues Element wird oben auf den Stapel gelegt.
2. **Pop**: Das oberste Element wird vom Stapel entfernt.
3. **Peek**: Das oberste Element wird betrachtet, bleibt aber auf dem Stapel.


Diese Operationen führen wir in einer Implementierung mithilfe einer verketteten Liste durch.
