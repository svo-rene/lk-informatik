# Queue (Schlange) mit einer Verketteten Liste

In der Informatik ist eine **Schlange** oder **Queue** eine Datenstruktur, die ähnlich wie eine Warteschlange im Alltag funktioniert: das erste Element, das hinzugefügt wurde, wird auch als erstes entfernt. Dies bezeichnet man als **FIFO (First In, First Out)**-Prinzip. Eine Queue wird häufig verwendet, wenn man Daten in einer Reihenfolge verarbeiten muss, z. B. beim Drucken von Dokumenten oder der Aufgabenverwaltung.

In dieser Lektion werden wir eine Queue mithilfe einer **verketteten Liste** implementieren und dabei auch das Prinzip FIFO (First In – First Out) betrachten.

### Eigenschaften einer Queue mit verketteter Liste
Eine Queue, die mit einer verketteten Liste implementiert ist, hat zwei Hauptoperationen:
- **Enqueue** (Hinzufügen): Ein Element wird am Ende der Liste hinzugefügt.
- **Dequeue** (Entfernen): Das vorderste Element wird aus der Liste entfernt.

### Einsatzbeispiele für Queues
1. **Warteschlangen in Geschäften**: Kunden treten der Warteschlange in der Reihenfolge ihres Eintreffens bei und werden in dieser Reihenfolge bedient.
2. **Druckerwarteschlange**: Dokumente, die an den Drucker gesendet werden, werden in der Reihenfolge des Eingangs gedruckt.
