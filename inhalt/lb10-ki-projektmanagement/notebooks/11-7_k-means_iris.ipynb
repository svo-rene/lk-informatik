```python
# Benötigte Bibliotheken importieren
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from sklearn import datasets
```

### Schritt 1: Den Iris-Datensatz laden und erkunden

Der Iris-Datensatz kann direkt von sklearn geladen werden.

```python
# Iris-Datensatz laden
iris = datasets.load_iris()
```

```python
# In DataFrame umwandeln
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

## Schritt 2: Datenvisualisierung

Wir beginnen mit einer einfachen Visualisierung, um die Verteilung der Daten besser zu verstehen.

```python
# Streudiagramm von Kelchblattlänge und Blütenblattlänge
plt.figure(figsize=(8, 6))
plt.scatter(df['sepal length (cm)'], df['petal length (cm)'], s=100)
plt.title('Kelchblattlänge vs. Blütenblattlänge')
plt.xlabel('Kelchblattlänge (cm)')
plt.ylabel('Blütenblattlänge (cm)')
plt.show()
```

## Schritt 3: Datenvorverarbeitung

Da K-Means empfindlich auf die Skalierung der Daten reagiert, standardisieren wir die Daten, um eine bessere Performance zu erreichen.

**Warum ist das nötig?** K-Means berechnet Abstände (z. B. den euklidischen Abstand, den du auf der Theorieseite kennengelernt hast), um Datenpunkte ihrem nächstgelegenen Cluster-Zentrum zuzuordnen. Wenn ein Merkmal viel größere Zahlenwerte hat als ein anderes, dominiert es automatisch die Abstandsberechnung – unabhängig davon, ob es inhaltlich wichtiger ist.

**Beispiel:** Stell dir vor, du hast die Merkmale "Bestellwert in Euro" (Werte zwischen 0 und 1000) und "Anzahl Bestellungen pro Jahr" (Werte zwischen 0 und 20). Bei der Berechnung des euklidischen Abstands würde der Bestellwert wegen seiner viel größeren Zahlenspanne fast allein über den Abstand entscheiden – die Anzahl der Bestellungen würde kaum ins Gewicht fallen. Durch die Standardisierung (Mittelwert = 0, Standardabweichung = 1) werden alle Merkmale auf eine vergleichbare Skala gebracht, sodass jedes Merkmal fair in die Abstandsberechnung eingeht.

```python
# Standardisierung der Daten
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

# Den standardisierten Datensatz anzeigen
print(pd.DataFrame(df_scaled, columns=df.columns).head())
```

## Schritt 4: K-Means-Clustering

Nun wenden wir den K-Means-Algorithmus an. Wir beginnen mit k=3, da wir wissen, dass der Iris-Datensatz 3 verschiedene Iris-Arten enthält.

```python
# K-Means mit 3 Clustern anwenden
kmeans = KMeans(n_clusters=3, n_init=10, random_state=42)
kmeans.fit(df_scaled)

# Cluster-Zuweisungen anzeigen
df['Cluster'] = kmeans.labels_

# Die Cluster-Zentren anzeigen
print("Cluster-Zentren:\n", kmeans.cluster_centers_)
```

### Aufgabe 4:
Erstelle ein Streudiagramm, das die Cluster visualisiert.

```python
# Streudiagramm der Cluster
plt.figure(figsize=(8, 6))
plt.scatter(df['sepal length (cm)'], df['petal length (cm)'], c=df['Cluster'], cmap='viridis', s=100)
plt.title('Kelchblattlänge vs. Blütenblattlänge mit Cluster-Zuweisungen')
plt.xlabel('Kelchblattlänge (cm)')
plt.ylabel('Blütenblattlänge (cm)')
plt.show()
```

## Schritt 5: Modellauswertung

K-Means-Clustering minimiert die sogenannte Trägheit (Inertia). Die Trägheit ist ein Maß für die Qualität der Clusterbildung im k-Means-Algorithmus. Sie gibt an, wie gut die Datenpunkte in den Clustern gruppiert sind.

Niedrige Trägheit: Dies bedeutet, dass die Datenpunkte nah an den Clusterzentren liegen und die Cluster gut definiert sind. Eine niedrige Trägheit zeigt, dass die Cluster kompakt und nicht zu weit verstreut sind.

Hohe Trägheit: Eine hohe Trägheit bedeutet, dass die Datenpunkte weit von ihren zugehörigen Clusterzentren entfernt sind, was darauf hindeutet, dass die Cluster möglicherweise nicht gut definiert oder stark verstreut sind.

**Hinweis:** Die Trägheit ist genau das, was auf der Theorieseite als SSE (Summe der quadrierten Abstände) bezeichnet wurde – nur der Name unterscheidet sich je nach Werkzeug. scikit-learn nennt diesen Wert `inertia_`.

```python
# Inertia berechnen
print(f"Trägheit des Modells: {kmeans.inertia_:.2f}")
```

## Schritt 6: Visualisierung mit PCA

Da der Iris-Datensatz 4 Dimensionen (Merkmale) hat, können wir ihn nicht direkt in einem normalen 2D-Diagramm darstellen – ein Streudiagramm hat schließlich nur eine x- und eine y-Achse. Deshalb verwenden wir die Hauptkomponentenanalyse (PCA, "Principal Component Analysis"), um die 4 Merkmale auf 2 neue Achsen zu reduzieren, die wir dann plotten können.

**Anschauliche Vorstellung:** Stell dir vor, du hältst ein Objekt (z. B. eine Kaffeetasse) ins Licht und wirfst damit einen Schatten an die Wand. Der 3D-Gegenstand wird dabei auf eine 2D-Fläche projiziert – dabei geht zwangsläufig etwas Information verloren, aber wenn du den Winkel geschickt wählst, erkennst du im Schatten trotzdem noch gut die Form der Tasse. Genau das macht PCA mit Daten: Sie sucht die "beste Blickrichtung" (die sogenannten Hauptkomponenten), aus der man die 4-dimensionalen Datenpunkte auf 2 Dimensionen projiziert, ohne dabei zu viele Unterschiede zwischen den Datenpunkten zu verlieren. Konkret sucht PCA dabei die Richtungen im Datenraum, in denen die Datenpunkte am stärksten streuen (variieren) – denn in diesen Richtungen steckt die meiste Information.

Wichtig: PCA verändert nicht die eigentlichen Cluster-Zuordnungen (die wurden ja schon vorher mit den originalen 4 Merkmalen berechnet) – sie dient hier ausschließlich der Visualisierung.

```python
# PCA anwenden
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_scaled)

# Streudiagramm der Cluster mit PCA
plt.figure(figsize=(8, 6))
plt.scatter(df_pca[:, 0], df_pca[:, 1], c=df['Cluster'], cmap='viridis', s=100)
plt.title('K-Means Clustering (2D PCA)')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.show()
```
