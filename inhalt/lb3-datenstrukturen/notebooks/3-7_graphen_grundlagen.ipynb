## Definition eines Graphen
Ein **Graph** besteht aus:
- Einer Menge von **Knoten** (Vertices).
- Einer Menge von **Kanten** (Edges), die die Verbindung zwischen den Knoten darstellen.

Graphen können ungerichtet, gerichtet und gewichtet sein. In diesem Notebook betrachten wir ungewichtete und ungerichtete Graphen.

```
# Klasse für einen Knoten
class Node:
    def __init__(self, value):
        self.value = value
        self.neighbors = []

    def add_neighbor(self, neighbor):
        self.neighbors.append(neighbor)

    def __repr__(self):
        return "Node(" + str(self.value) + ")"

# Klasse für einen Graphen
class Graph:
    def __init__(self):
        self.nodes = {}

    def add_node(self, value):
        if value not in self.nodes:
            self.nodes[value] = Node(value)

    def add_edge(self, value1, value2):
        if value1 in self.nodes and value2 in self.nodes:
            self.nodes[value1].add_neighbor(self.nodes[value2])
            self.nodes[value2].add_neighbor(self.nodes[value1])  # Ungerichteter Graph

    def __repr__(self):
        text = ""
        for value in self.nodes:
            nachbarn = []
            for n in self.nodes[value].neighbors:
                nachbarn.append(n.value)
            text = text + str(value) + ": " + str(nachbarn) + "\n"
        return text
```

## Beispiel: Erstellen eines Graphen
Wir erstellen einen Graphen mit den Knoten `A`, `B`, `C`, `D` und den Kanten `A-B`, `A-C`, `B-D`, `C-D`. Dabei nutzen wir die oben definierten Klassen.

```
# Graph erstellen
g = Graph()
g.add_node("A")
g.add_node("B")
g.add_node("C")
g.add_node("D")

g.add_edge("A", "B")
g.add_edge("A", "C")
g.add_edge("B", "D")
g.add_edge("C", "D")

print(g)
```

## Breitensuche (Breadth-First Search, BFS)
Die Breitensuche arbeitet Ebene für Ebene und durchsucht alle benachbarten Knoten, bevor sie tiefer in den Graphen geht. Dazu brauchen wir eine Schlange (Queue). Wir nehmen die **Queue mit Knoten aus Seite 3.2**: `enqueue` hinten, `dequeue` vorn – beides in O(1). Eine Python-Liste mit `pop(0)` würde auch funktionieren, ließe aber bei jedem Entnehmen alle übrigen Einträge nachrücken (O(n)).

Hast du `queue_knoten.py` aus 3.2 im selben Ordner, genügt `from queue_knoten import Queue`. Damit das Notebook auch allein läuft, steht die Klasse hier noch einmal. Ihr Knoten heißt `QueueNode`, weil `Node` in diesem Notebook schon der Graph-Knoten ist.

```
# Queue mit Knoten aus Seite 3.2 (wie in queue_knoten.py)
class QueueNode:
    def __init__(self, data):
        self.data = data
        self.next = None

class Queue:
    def __init__(self):
        self.__head = None    # vorn: hier wird entnommen
        self.__tail = None    # hinten: hier wird eingefügt

    def is_empty(self):
        return self.__head is None

    def enqueue(self, data):
        neu = QueueNode(data)
        if self.__tail is None:         # Schlange war leer
            self.__head = neu
        else:
            self.__tail.next = neu
        self.__tail = neu

    def dequeue(self):
        if self.is_empty():
            raise IndexError("dequeue aus leerer Schlange")
        data = self.__head.data
        self.__head = self.__head.next
        if self.__head is None:         # letztes Element entnommen
            self.__tail = None
        return data
```

```
# Implementierung der Breitensuche
def bfs(graph, start_value):
    visited = [start_value]          # schon entdeckte Knoten
    queue = Queue()
    queue.enqueue(start_value)

    while not queue.is_empty():
        current = queue.dequeue()
        print(current, end=" ")

        for neighbor in graph.nodes[current].neighbors:
            if neighbor.value not in visited:
                visited.append(neighbor.value)
                queue.enqueue(neighbor.value)

# BFS ausführen
bfs(g, "A")
```

## Tiefensuche (Depth-First Search, DFS)
Die Tiefensuche arbeitet rekursiv oder iterativ und untersucht einen Pfad im Graphen, bevor sie andere Pfade untersucht. Hier verwenden wir die rekursive Variante.

```
# Implementierung der Tiefensuche (rekursiv)
def dfs(graph, start_value, visited):
    visited.append(start_value)
    print(start_value, end=" ")

    for neighbor in graph.nodes[start_value].neighbors:
        if neighbor.value not in visited:
            dfs(graph, neighbor.value, visited)

# DFS ausführen – visited ist am Anfang eine leere Liste
dfs(g, "A", [])
```
