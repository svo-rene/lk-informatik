# Binäre Suchbäume

Ein **Binärer Suchbaum** ist ein spezieller Typ eines Binärbaums, bei dem für jeden Knoten gilt:
1. Alle Werte im **linken Teilbaum** sind kleiner als der Wert des Knotens.
2. Alle Werte im **rechten Teilbaum** sind größer als der Wert des Knotens.

Diese Eigenschaft macht den Binären Suchbaum ideal für Operationen wie das effiziente Suchen, Einfügen und Löschen von Elementen.

### Beispiel eines Binären Suchbaums:
```
        10
       /  \
      6    15
     / \     \
    3   8    20
```

Inorder-Traversierung eines Binären Suchbaums ergibt immer eine sortierte Reihenfolge der Werte.

- **Tiefensuche:** Gehe so tief wie möglich in einem Teilbaum, bevor du andere Teilbäume erkundest. Hierbei werden Knoten über Preorder, Inorder oder Postorder-Traversierungen besucht.
- **Breitensuche (Level Order):** Besuche die Knoten Ebene für Ebene, bevor du tiefer in den Baum gehst.
