# Vergleich von Array und Linked List

### 1. Grundlegender Unterschied zwischen Array und Linked List

Ein **Array** ist eine Datenstruktur, in der Elemente in einem zusammenhängenden Speicherbereich gespeichert werden. Alle Elemente haben denselben Datentyp und sind über einen Index erreichbar. Eine **Linked List** hingegen besteht aus einer Verkettung von Knoten, wobei jeder Knoten sowohl die Daten als auch einen Zeiger auf den nächsten Knoten enthält. Diese Knoten sind im Speicher nicht notwendigerweise zusammenhängend, sondern durch Zeiger miteinander verbunden.

**Array:**

- Feste Größe, die beim Erstellen des Arrays festgelegt wird.
- Direkter Zugriff auf die Elemente durch einen Index.
- Beispiel: Ein Array mit den Zahlen `[5, 10, 15, 20]` hat festen Speicherbedarf und die Werte sind aufeinanderfolgend im Speicher abgelegt.

**Linked List:**

- Dynamische Größe, kann während der Laufzeit wachsen oder schrumpfen.
- Kein direkter Zugriff auf Elemente – man muss die Liste ab dem ersten Element durchlaufen, um an eine bestimmte Position zu gelangen.
- Beispiel: Eine Linked List, die dieselben Zahlen `5 -> 10 -> 15 -> 20` enthält, benötigt zusätzlichen Speicher für die Zeiger und die Knoten müssen nicht zusammenhängend gespeichert sein.
- Jeder Knoten kann mehrere Informationen enthalten. 

### 2. Vorteile und Nachteile einer Linked List

#### **Vorteile von Linked Lists**

1. **Dynamische Speicherzuweisung**: Die Größe einer Linked List kann dynamisch zur Laufzeit verändert werden. Das ist besonders hilfreich, wenn man nicht weiß, wie viele Elemente man benötigen wird.
- **Beispiel**: Wenn du eine Liste von Studenten speichern musst, deren Anzahl sich ändert (weil neue Studenten hinzukommen oder andere abgehen), dann ist eine Linked List flexibler als ein Array, das eine feste Länge hat.

2. **Einfache Speicherung von komplexen Daten**: Jeder Knoten kann eine beliebige Anzahl von Informationen enthalten. Das macht Linked Lists ideal für komplexere Datenstrukturen, wie z. B. in einer Bibliotheksverwaltung.
- **Beispiel**: Für eine Bibliotheksverwaltung kann jeder Knoten alle relevanten Informationen eines Buches enthalten: **Titel, Autor, Seitenzahl, ISBN**. Diese Flexibilität erleichtert es, komplexe Datensätze zu speichern und zu verwalten.

3. **Einfügen und Löschen**: Das Einfügen oder Löschen von Elementen ist bei Linked Lists effizient, da man nur die Zeiger der benachbarten Knoten aktualisieren muss, ohne alle folgenden Elemente wie in einem Array verschieben zu müssen.
- **Beispiel**: Wenn ein neues Element in der Mitte der Liste eingefügt werden soll, kann dies bei einer Linked List einfach durch eine Zeigeraktualisierung geschehen. Bei einem Array müssten alle nachfolgenden Elemente verschoben werden.

#### **Nachteile von Linked Lists**

1. **Kein direkter Zugriff**: Bei einer Linked List ist der Zugriff auf Elemente langsamer, da man die Liste Knoten für Knoten durchlaufen muss.
- **Beispiel**: Wenn du auf das 50. Element einer Liste zugreifen möchtest, musst du bei einer Linked List von Knoten 1 bis 49 durchgehen. Bei einem Array kannst du direkt auf das 50. Element zugreifen.

2. **Zusätzlicher Speicherbedarf**: Jeder Knoten benötigt zusätzlichen Speicher für den Zeiger. Bei vielen Knoten führt dies zu einem größeren Speicherverbrauch als bei Arrays.
- **Beispiel**: Ein Linked List mit 100 Elementen benötigt 100 zusätzliche Speicherstellen für die Zeiger, was besonders bei beschränktem Speicher ineffizient sein kann.

### 3. Aufgaben: Wähle begründet welche ob du in den folgenden Fällen ein Array oder eine Linked List verwenden würdest


1. **Aufgabe 1:** Du sollst eine Liste von Temperaturen an jedem Tag eines Monats speichern. Die Anzahl der Tage ist bekannt und fix (30 oder 31).


2. **Aufgabe 2:** Du möchtest eine Teilnehmerliste für eine Veranstaltung erstellen. Teilnehmer können sich jederzeit an- oder abmelden, und du weißt nicht im Voraus, wie viele Teilnehmer es geben wird.


3. **Aufgabe 3:** Ein Programm benötigt eine Liste mit den ersten 1.000 natürlichen Zahlen. Du wirst diese Liste oft durchsuchen und auf bestimmte Positionen zugreifen.


4. **Aufgabe 4:** Du baust eine To-Do-Liste, bei der es möglich sein soll, Aufgaben flexibel hinzuzufügen oder zu entfernen, auch an bestimmten Positionen in der Liste.


5. **Aufgabe 5:** Du erstellst ein Bibliotheksverwaltungssystem, bei dem jedes Buch Informationen wie Titel, Autor, Seitenzahl und ISBN enthält. Diese Bücher können jederzeit hinzugefügt oder entfernt werden.
