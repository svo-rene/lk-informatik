# Übungsaufgaben

## Aufgabe 1
Implementiere die Schlange aus den Kopf. Notiere dir zunächst alle wichtigen Vorüberlegungen und zeiche dir Datenstruktur Schlange, bevor du beginnst.

 Stell dir vor, du stehst in einer Warteschlange an der Supermarktkasse. Der Kunde, der zuerst kommt, wird auch zuerst bedient.
 Implementiere eine Schlange (FIFO) mithilfe einer verketteten Liste.
 - Die Schlange soll die folgenden Methoden enthalten:
   1. enqueue(value): Fügt einen Kunden am Ende der Warteschlange hinzu.
   2. dequeue(): Entfernt den vordersten Kunden aus der Warteschlange und gibt ihn zurück.
   3. wartezeit_bestimmen(average): berechnet ausgehend von einer average Zeit an wie lange das letzte Element in der Schlange warten muss



## Aufgabe 2: Klammerprüfung mit einem Stapel 

Stell dir vor, du machst deine Mathematikhausaufgaben und musst prüfen, ob alle Klammern korrekt gesetzt sind.
Implementiere eine Funktion, die prüft, ob die Klammern in einem gegebenen mathematischen Ausdruck korrekt geschlossen wurden.
 - Verwende einen Stapel, um die Klammern zu verfolgen. Implementiere den Stapel und schreibe anschließend eine Methode klammer_pruefen()
 - Beispiel: "(a + b) * (c - d)" sollte als korrekt erkannt werden.

## Aufgabe 3: Minimum-Element in einem Stapel finden (Temperaturen überwachen)

 Stell dir vor, du misst täglich die Temperatur und speicherst sie in einem Stapel.
 Implementiere eine Methode, die die niedrigste Temperatur im Stapel findet.
 - Beispiel: Wenn der Stapel die Werte [3, 5, 2, 1, 4] enthält, sollte die Methode 1 zurückgeben.

## Aufgabe 4: Rückwärtslesen mit einem Stapel (Nachricht umdrehen)

 Stell dir vor, du möchtest eine geheime Nachricht verschlüsseln, indem du sie rückwärts liest.
 Implementiere eine Funktion, die eine Zeichenfolge umkehrt.
 - Verwende dazu einen Stapel, um die Zeichen der Zeichenfolge in umgekehrter Reihenfolge zu speichern.
