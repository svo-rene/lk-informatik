## 1. Divide and Conquer
**Wann wird es verwendet?**
Divide and Conquer wird verwendet, wenn ein Problem durch das Zerlegen in kleinere, gleichartige Teilprobleme effizienter gelöst werden kann. Wenn sich das Problem rekursiv in kleinere Unterprobleme zerlegen und deren Lösungen zusammenfügen lassen, ist dies eine gute Strategie.

**Anwendungsfälle:**
- **Binäre Suche** (schnelle Suche in sortierten Daten)
- **Mergesort** (effiziente Sortierung)
- **Exponentiation by Squaring** (schnelle Potenzberechnung)
- **Fast Fourier Transform (FFT)** (Signalverarbeitung, Bildverarbeitung)

### Beispiel: Binäre Suche (Pseudocode)
Die Binäre Suche ist ein Divide-and-Conquer-Algorithmus, der in einer **sortierten** Liste nach einem Element sucht. Dabei wird die Liste immer wieder halbiert, bis das gesuchte Element gefunden oder ausgeschlossen wurde.
```
Funktion binäre_suche(liste, links, rechts, ziel):
    falls links > rechts:
        Rückgabe -1
    mitte = (links + rechts) // 2
    falls liste[mitte] == ziel:
        Rückgabe mitte
    falls liste[mitte] > ziel:
        Rückgabe binäre_suche(liste, links, mitte - 1, ziel)
    sonst:
        Rückgabe binäre_suche(liste, mitte + 1, rechts, ziel)
```

````python
def binary_search(arr, left, right, target):
    if left > right:
        return -1
    mid = (left + right) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return binary_search(arr, left, mid - 1, target)
    else:
        return binary_search(arr, mid + 1, right, target)

# Test
arr = [1, 3, 5, 7, 9, 11, 13]
print(binary_search(arr, 0, len(arr)-1, 9))  # Erwartete Ausgabe: 4
````

### Mergesort

````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])

    return merge(left_half, right_half)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Test
arr = [38, 27, 43, 3, 9, 82, 10]
print(merge_sort(arr))  # Erwartete Ausgabe: [3, 9, 10, 27, 38, 43, 82]
````

## 2. Traversierung
**Wann wird es verwendet?**
Traversierungsalgorithmen werden in Datenstrukturen wie **Bäume und Graphen** eingesetzt, um deren Elemente systematisch zu durchlaufen.

**Anwendungsfälle:**
- **Web-Crawling** (Breitensuche BFS für Suchmaschinen)
- **Suche und Ausgabe von Bäumen** (Pre-, In-, Poster-Order)
- **Navigation in Straßenkarten** (Dijkstra nutzt BFS)
- **Spiel-KI** (Tiefensuche DFS für Spielbäume)

````python
def dfs(graph, node, visited):
    if node not in visited:  
        visited.append(node) 
        print(node, end=' ')
        for neighbor in graph[node]:
            dfs(graph, neighbor, visited)

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [], 
    'E': [], 
    'F': []
}

# Test mit Liste
dfs(graph, 'A', []) 
````

## 3. Backtracking
**Wann wird es verwendet?**
Backtracking wird verwendet, wenn man eine Lösung schrittweise aufbaut und dabei Entscheidungen zurücknehmen kann, falls ein gewählter Weg nicht zu einer Lösung führt.

**Anwendungsfälle:**
- **Sudoku-Löser**
- **Routenplanung mit Hindernissen**
- **Das N-Damen-Problem** (Schach)

````python
def find_combinations(coins, target, combination=[], index=0):
    if target == 0:
        print(combination)
        return
    if target < 0 or index >= len(coins):
        return
    find_combinations(coins, target - coins[index], combination + [coins[index]], index)
    find_combinations(coins, target, combination, index + 1)

# Test
coins = [1, 2, 5]
target = 5
find_combinations(coins, target)
````
