# Datenstruktur: Bäume

### Definition
Ein Baum ist eine hierarchische Datenstruktur, die aus **Knoten (Nodes)** besteht. Ein spezieller Knoten, die sogenannte **Wurzel (Root)**, ist der Ausgangspunkt des Baumes. Jeder Knoten kann Kinderknoten (Child Nodes) haben und genau einen Elternknoten (Parent Node), außer der Wurzel.<br>
Bäume kombinieren die Vorteile von Arrays (effiziente Suche bei sortierten Daten) und verketteten Listen (dynamische Speicherverwaltung) und erweitern diese durch hierarchische Organisation

**Eigenschaften von Bäumen:**
1. Es gibt genau einen Wurzelknoten.
2. Jeder Knoten hat genau einen Elternknoten (außer der Wurzel).
3. Es gibt keine Zyklen.
4. Jeder Knoten kann beliebig viele Kinder haben (abhängig vom Baumtyp).

**Begriffe:**
- **Wurzel**: Der oberste Knoten im Baum.
- **Blatt**: Knoten ohne Kinder (unterste Ebene)
- **Kind**: Nachfolger eines Knotens
- **Eltern**: Vorgänger eines bestimmten Knotens
- **Teilbaum**: Ein Teil des Baumes, der selbst ein Baum ist.
- **Höhe eines Baums**: Die maximale Anzahl von Kanten von der Wurzel zu einem Blatt.
- **Tiefe eines Knotens**: Die Anzahl von Kanten von der Wurzel zu diesem Knoten.

### Anwendung
- Dateisysteme: Darstellung von Ordner- und Datei-Hierarchien (z. B. in Betriebssystemen)
- Suchalgorithmen und Datenbanken: Binäre Suchbäume (BST) für effizientes Suchen und B-Bäume für große Datenbanken.
- Entscheidungsbäume in der Künstlichen Intelligenz: Klassifikation und Regression in maschinellem Lernen.
- Huffman-Bäume für Datenkompression: Kodierung von Daten zur verlustfreien Kompression (z. B. ZIP oder JPEG).
- Webtechnologien: Darstellung und Manipulation der Struktur von Webseiten (HTML, XML)."
