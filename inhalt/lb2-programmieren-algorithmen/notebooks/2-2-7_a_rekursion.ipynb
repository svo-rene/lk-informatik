# Rekursion



## Iteration vs. Rekursion

**Iteration:**
Ein iterativer Algorithmus führt eine Aktion so lange wiederholt aus, bis eine Abbruchbedingung erfüllt ist. Beispiele aus dem Alltag:
- Spülen eines Tellerstapels: Nimm einen Teller, spüle ihn ab, lege ihn auf den Stapel mit sauberen Tellern und wiederhole, bis keine schmutzigen Teller mehr da sind.

**Rekursion:**
Ein rekursiver Algorithmus ruft sich selbst auf, um das Problem in kleinere Teile zu zerlegen, bis ein Basisfall erreicht ist. 
- Beispiel: Eine Matroschka-Puppe: Öffne die äußere Puppe, arbeite mit der nächstkleineren Puppe genauso weiter, bis du zur kleinsten Puppe kommst (Basisfall).

| Merkmal | Iteration | Rekursion |
|---------|-----------|-----------|
| Wiederholung | Schleifen (z.B. `for`, `while`) | Selbstaufruf einer Funktion |
| Abbruch | Explizite Bedingung (z.B. `while`) | Basisfall in der Funktion |
| Effizienz | Speicher effizienter | Kann höheren Speicherbedarf haben (Rekursionsstapel) |

### Beispiel: Matroschka


```python
#iterativ:
def mat_iter(n):
    for i in range(n):
        print("Packe Matroschka aus. Die Variable i beträgt:", i)
    print("Matroschka fertig ausgepackt.")


#rekursiv:
def mat_reku(n):
    if n == 0:
        print("Matroschka fertig ausgepackt.")
    else:
        print("Packe Matroschka aus. Die Variable n beträgt:", n)
        return mat_reku(n-1)


print("iterativ:")
mat_iter(5)

print("und jetzt rekursiv:")
mat_reku(5)
```

## Endrekursion und rekursiver Abstieg

Das Beispiel der Matroschka ist ein Beispiel für eine endrekursive Funktion. Bei endrekursiven Funktionen steht das Ergebnis bereits nach dem Rekursionabstieg fest, sobald die Abbruchbedinung erreicht ist.

- **Rekursiver Abstieg:** Der Prozess, bei dem die Funktion sich selbst wiederholt aufruft, um das Problem in kleinere Teile zu zerlegen.
- **Abbruchbedingung:** Die Abbruchbedingung ist erreicht, sobald der Basisfall im "if-Teil" der Verzweigung erreicht ist. Die Abbruchbedingung beschreibt meist das kleinste, zerlegte Teilproblem des Gesamtproblems. Im Beispiel der Matroschka war es "Du bist fertig, sobald es keine Matroschka mehr in der Matroschka gibt --> n == 0 .

### Beispiel 2: Summe der Zahlen bis `n`

Es sollen die Summe der Zahlen von 0 bis n berechnet werden.

```python
def sum_iter(n):
    summe = 0
    for i in range(n+1):
        summe = summe + i
        print("Zwischenergebnis:", summe)
    return summe

def sum_endreku(n, summe = 0):
    print("summe_endreku: n =", n, "summe =", summe)
    if n == 0:  
        return summe
    return sum_endreku(n - 1, summe + n)

def sum_echtreku(n):
    print("sum_echtreku: n =", n)
    if n == 0:  
        return 0
    result = n + sum_echtreku(n - 1)  # Rekursiver Aufruf
    print("Rückkehr: n =", n, "Teilsumme =", result)
    return result

print("iterativ:")
print(sum_iter(5))

print("und jetzt endrekursiv:")
print(sum_endreku(5))  # Ausgabe: 15

print("und jetzt echtrekursiv:")
print(sum_echtreku(5))
```

# Echtrekursion und rekursiver Aufstieg

Beim echtrekursiven Funktionen erfolgt zuerst ein rekursiver Abstieg, bei welcher die Funktion sich immer wieder mit kleiner werdenden Werten aufruft bis die Abbruchbedingung erreicht ist, danach erfolgt ein rekursiver Aufstieg, bei welchem die Zwischenwerte an die nächste Funktion weitergegeben werden, bis der Funktionsstapel abgearbeitet ist und das Endergebnis berechnet ist.

- **Rekursiver Aufstieg:** Nach dem Basisfall kehren die Funktionsaufrufe in umgekehrter Reihenfolge zurück. Während des Rücksprungs werden die Ergebnisse der Teilprobleme kombiniert.

In dem Summe bis n Beispiel, werden nach der Abbruchbedingung n == 0 die Teilergebnisse der Summen zurückgegeben und mit dem derzeitigen n addiert. Wenn man alle Teilprobleme aufaddiert, erhält man das Endergebnis.
