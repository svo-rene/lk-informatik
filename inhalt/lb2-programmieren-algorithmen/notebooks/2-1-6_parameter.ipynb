
# Werteparameter und Referenzparameter in Python

In Python gibt es einen Unterschied zwischen der Übergabe von Werten und der Übergabe von Referenzen an Funktionen. Dieser Unterschied ist wichtig, um zu verstehen, wie sich Variablen in Funktionen verhalten.

In Python werden unveränderliche Objekte (wie Zahlen und Strings) als Werteparameter übergeben, während veränderliche Objekte (wie Listen und Dictionaries) als Referenzparameter übergeben werden.

- Werteparameter: Änderungen innerhalb der Funktion beeinflussen nicht die ursprüngliche Variable
- Referenzparameter: Änderungen innerhalb der Funktion beeinflussen das ursprüngliche Objekt


### Themen:
1. Werteparameter (Call by Value)
2. Referenzparameter (Call by Reference)
3. Beispiele und Übungen



## 1. Werteparameter (Call by Value)

Wenn eine Variable als Werteparameter an eine Funktion übergeben wird, wird eine Kopie des Werts erstellt. Das bedeutet, dass Änderungen an der Variable innerhalb der Funktion nicht die Originalvariable außerhalb der Funktion beeinflussen.

In diesem Beispiel bleibt der Wert von `x` außerhalb der Funktion unverändert.


```python
def ändere_zahl(zahl):
    zahl = 10
    print("Innerhalb der Funktion:", zahl)

x = 5
ändere_zahl(x)
print("Außerhalb der Funktion:", x)
```


## 2. Referenzparameter (Call by Reference)

Wenn eine Variable als Referenzparameter übergeben wird, wird keine Kopie des Werts erstellt, sondern die ursprüngliche Referenz verwendet. Das bedeutet, dass Änderungen innerhalb der Funktion auch die Originalvariable außerhalb der Funktion beeinflussen können.

In diesem Fall wird die Liste sowohl innerhalb als auch außerhalb der Funktion verändert.


```python
def ändere_liste(liste):
    liste.append(4)
    print("Innerhalb der Funktion:", liste)

meine_liste = [1, 2, 3]
ändere_liste(meine_liste)
print("Außerhalb der Funktion:", meine_liste)
```
