# Werteparameter und Referenzparameter in Python

Viele Programmiersprachen unterscheiden zwei Arten von Parametern:

- **Werteparameter** (call by value): Die Funktion bekommt eine Kopie des Werts. Änderungen bleiben in der Funktion.
- **Referenzparameter** (call by reference): Die Funktion bekommt einen Verweis auf die Variable des Aufrufers. Jede Änderung, auch eine neue Zuweisung, wirkt nach außen.

**Python kennt keine echten Referenzparameter.** Python übergibt immer eine Referenz auf das Objekt, kopiert wird dabei nichts (*call by sharing*). Ob der Aufrufer eine Änderung sieht, hängt davon ab, was die Funktion mit dem Parameter macht:

- Bekommt der Parameter einen **neuen Wert zugewiesen**, zeigt nur der Parameter auf ein neues Objekt. Die Variable des Aufrufers bleibt unverändert. Das wirkt wie ein Werteparameter.
- Wird ein **veränderbares Objekt** (Liste, Dictionary) über den Parameter **verändert**, sieht der Aufrufer die Änderung, denn beide Namen zeigen auf dasselbe Objekt. Das wirkt wie ein Referenzparameter.

### Themen:
1. Neue Zuweisung im Unterprogramm (wirkt wie ein Werteparameter)
2. Liste im Unterprogramm verändern (wirkt wie ein Referenzparameter)
3. Beispiele und Übungen

## 1. Neue Zuweisung im Unterprogramm (wirkt wie ein Werteparameter)

Beim Aufruf zeigt der Parameter `zahl` auf dasselbe Objekt wie `x`, die 5. Die Zuweisung `zahl = 10` lässt nur `zahl` auf ein neues Objekt zeigen, `x` zeigt weiter auf die 5. Kopiert wird dabei nichts.

In diesem Beispiel bleibt der Wert von `x` außerhalb der Funktion unverändert.

```python
def ändere_zahl(zahl):
    zahl = 10
    print("Innerhalb der Funktion:", zahl)

x = 5
ändere_zahl(x)
print("Außerhalb der Funktion:", x)
```

## 2. Liste im Unterprogramm verändern (wirkt wie ein Referenzparameter)

`liste` und `meine_liste` zeigen auf **dieselbe** Liste. `liste.append(4)` verändert dieses eine Objekt, deshalb sieht der Aufrufer die Änderung.

Achtung: Eine neue Zuweisung wie `liste = [9]` in der Funktion würde `meine_liste` nicht verändern. Python hat eben keine echten Referenzparameter.

In diesem Fall wird die Liste sowohl innerhalb als auch außerhalb der Funktion verändert.

```python
def ändere_liste(liste):
    liste.append(4)
    print("Innerhalb der Funktion:", liste)

meine_liste = [1, 2, 3]
ändere_liste(meine_liste)
print("Außerhalb der Funktion:", meine_liste)
```
