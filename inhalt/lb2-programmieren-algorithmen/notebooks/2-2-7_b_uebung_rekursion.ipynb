## Zusatzaufgabe
Beispiel: Zahlenfolge 5, 10, 20, 40, 80, ...


Die Folge ist rekursiv definiert als:

- **Startwert**: $ a_1 = 5 $
- **Rekursionsregel**: $a_n = 2 \cdot a_{n-1}$

Die iterative Darstellung lautet:

- $ a_n = 5 \cdot 2^{n-1} $
    

```python

# Rekursive Berechnung der n-ten Zahl der Folge
def rekursive_folge(n):
    if n == 1:
        return 5
    else:
        return 2 * rekursive_folge(n - 1)

# Iterative Berechnung der n-ten Zahl der Folge
def iterative_folge(n):
    return 5 * (2 ** (n - 1))

# Test für n = 1, 2, 3, 4, 5, 6
print("Rekursiv:", rekursive_folge(1), rekursive_folge(2), rekursive_folge(3), rekursive_folge(4), rekursive_folge(5), rekursive_folge(6))
print("Iterativ:", iterative_folge(1), iterative_folge(2), iterative_folge(3), iterative_folge(4), iterative_folge(5), iterative_folge(6))
    
```
