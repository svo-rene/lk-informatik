## 2. Schleifen (`while`, `for`)

Schleifen ermöglichen es, einen Codeblock mehrfach auszuführen, solange eine Bedingung erfüllt ist.

### `while`-Schleife:

Eine `while`-Schleife führt den Code so lange aus, wie die Bedingung `True` ist.


```python
zähler = 0
while zähler < 5:
    print("Zähler ist:", zähler)
    zähler = zähler + 1
```

### `for`-Schleife:

Die `for`-Schleife wird verwendet, um über eine Sequenz (z.B. Liste, Zeichenkette) zu iterieren.


```python
früchte = ["Apfel", "Banane", "Kirsche"]
for frucht in früchte:
    print(frucht)
```

### `for`-Schleife mit `range()`:

Die `range()`-Funktion wird verwendet, um eine Sequenz von Zahlen zu generieren, die in der `for`-Schleife durchlaufen werden kann.

**Parameter von `range()`:**

- `range(stop)`: Erzeugt eine Sequenz von `0` bis `stop` (exklusiv).
- `range(start, stop)`: Erzeugt eine Sequenz von `start` (inklusive) bis `stop` (exklusiv).
- `range(start, stop, step)`: Erzeugt eine Sequenz von `start` bis `stop` mit Schrittweite `step` (Schrittweite kann auch negativ sein)



```python
for i in range(5):
    print(i)
```

## 3. Verzweigungen (`if`, `else`, `elif`)

Verzweigungen werden verwendet, um verschiedene Codeabschnitte basierend auf Bedingungen auszuführen. Die gängigsten Verzweigungen in Python sind `if`, `elif` und `else`.


```python
alter = 18

if alter < 18:
    print("Du bist minderjährig.")
elif alter == 18:
    print("Du bist genau 18!")
else:
    print("Du bist volljährig.")
```
