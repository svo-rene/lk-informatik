




# Key-Value Datenbanken

Eine Key-Value-Datenbank ist eine einfache Form einer Datenbank, die Daten als Paare von Schlüssel (Key) und Wert (Value) speichert. Dies ist ähnlich wie bei einem Dictionary in Python. Jeder Schlüssel ist eindeutig und kann verwendet werden, um einen zugehörigen Wert abzurufen oder zu setzen.

**Merke:** Eine Key-Value-Datenbank ist besonders nützlich, wenn schnelle Zugriffe auf einfache Datenstrukturen benötigt werden, zum Beispiel bei Konfigurationen oder Caches.

## Was ist ein Python-Dictionary?

Ein **Dictionary** in Python ist eine Datenstruktur, die eine Sammlung von Schlüssel-Wert-Paaren speichert. Jeder Schlüssel ist eindeutig und wird verwendet, um auf einen entsprechenden Wert zuzugreifen. Dies funktioniert ähnlich wie ein echtes Wörterbuch, bei dem man ein Wort (den Schlüssel) nachschlägt und dessen Definition (den Wert) erhält.

Ein Dictionary in Python wird mit geschweiften Klammern `{ }` erstellt, und die Daten werden als Paare von Schlüssel und Wert darin abgelegt.

````python
# Ein einfaches Dictionary
benutzer = {
    'Name': 'Alice',
    'Alter': 25,
    'Beruf': 'Entwicklerin'
}

# Zugriff auf Werte über den Schlüssel
print("Name:", benutzer['Name'])
print("Alter:", benutzer['Alter'])

# Wert ändern
benutzer['Alter'] = 26
print("Aktualisiertes Alter:", benutzer['Alter'])

# Ein neues Schlüssel-Wert-Paar hinzufügen
benutzer['Stadt'] = 'Berlin'
print("Stadt hinzugefügt:", benutzer['Stadt'])

# Schlüssel-Wert-Paar löschen
del benutzer['Beruf']
print("Beruf entfernt:", benutzer)
````

> **Funktionen Dictionary:** Python-Dictionaries bieten viele nützliche Funktionen, um mit Key-Value-Daten zu arbeiten. Hier sind die wichtigsten:
>
> - `get()`: Diese Funktion ruft den Wert zu einem bestimmten Schlüssel ab. Sie ist nützlich, um sicherzustellen, dass kein Fehler auftritt, wenn der Schlüssel nicht existiert.
> - `keys()`: Diese Funktion gibt alle Schlüssel im Dictionary zurück.
> - `values()`: Diese Funktion gibt alle Werte im Dictionary zurück, ohne die zugehörigen Schlüssel.
> - `items()`: Diese Funktion gibt alle Schlüssel-Wert-Paare gleichzeitig zurück – besonders praktisch zum Durchlaufen (Iterieren) eines Dictionarys.
> - `len()`: Gibt die Anzahl der gespeicherten Schlüssel-Wert-Paare zurück.

````python
wert = benutzer.get('Name', 'Unbekannt')
print(wert)

alle_schluessel = list(benutzer.keys())
print(alle_schluessel)

alle_werte = list(benutzer.values())
print(alle_werte)

anzahl = len(benutzer)
print(anzahl)
````

### Über ein Dictionary iterieren

Mit einer `for`-Schleife kannst du alle Schlüssel eines Dictionarys der Reihe nach durchgehen und darüber auf die zugehörigen Werte zugreifen. Mit `.items()` bekommst du Schlüssel und Wert direkt zusammen. Das ist genau das Muster, das du gleich in den Lösungen der Übungsaufgaben wiederfindest.

````python
for name in benutzer:
    print(name, benutzer[name])

print()

for schluessel, wert in benutzer.items():
    print(schluessel, "->", wert)
````

### Dictionaries vs. Listen: Wo liegt der Vorteil?

Eine Liste eignet sich gut, wenn du gleichartige Elemente in einer bestimmten Reihenfolge speichern willst (z. B. alle Namen einer Klasse). Willst du aber zu einem eindeutigen Namen gezielt einen Wert nachschlagen, ist ein Dictionary die bessere Wahl:

````python
# Mit einer Liste musst du dir die Reihenfolge merken:
benutzer_liste = ["Alice", 25, "Entwicklerin"]
print(benutzer_liste[1])  # 25 - aber was war an Position 1 nochmal?

# Mit einem Dictionary greifst du über einen aussagekräftigen Namen zu:
benutzer_dict = {"Name": "Alice", "Alter": 25, "Beruf": "Entwicklerin"}
print(benutzer_dict["Alter"])  # 25 - sofort klar, was gemeint ist
````

**Merke:** Nutze eine Liste für geordnete Sammlungen gleichartiger Elemente. Nutze ein Dictionary, wenn du Werte über einen eindeutigen, aussagekräftigen Schlüssel nachschlagen willst – das ist genau das Prinzip einer Key-Value-Datenbank.

## Verschachtelte Dictionaries

Ein **verschachteltes Dictionary** ist ein Dictionary, bei dem die Werte selbst wieder Dictionaries sein können. Dies ist nützlich, wenn man komplexere Datenstrukturen wie mehrere Benutzer und deren jeweilige Informationen speichern möchte.

````python
# Ein Dictionary, das mehrere Benutzer speichert
benutzer_db = {
    'Alice': {'Alter': 25, 'Beruf': 'Entwicklerin'},
    'Bob': {'Alter': 30, 'Beruf': 'Designer'},
    'Clara': {'Alter': 22, 'Beruf': 'Lehrerin'}
}

# Zugriff auf die Daten eines Benutzers
print("Alices Alter:", benutzer_db['Alice']['Alter'])
print("Bobs Beruf:", benutzer_db['Bob']['Beruf'])

# Alter von Clara ändern
benutzer_db['Clara']['Alter'] = 23
print("Claras aktualisiertes Alter:", benutzer_db['Clara']['Alter'])

# Einen neuen Benutzer hinzufügen
benutzer_db['David'] = {'Alter': 28, 'Beruf': 'Ingenieur'}
print("Neuer Benutzer hinzugefügt:", benutzer_db['David'])

# Beruf von Bob löschen
del benutzer_db['Bob']['Beruf']
print("Bobs Beruf gelöscht:", benutzer_db['Bob'])
````

## Ausblick: Key-Value-Datenbanken in der Praxis und JSON

In der Realität werden Key-Value-Datenbanken häufig in groß angelegten Anwendungen verwendet. Sie arbeiten oft mit **JSON** (JavaScript Object Notation), einem Datenformat, das der Struktur von Python-Dictionaries sehr ähnlich ist. Viele moderne NoSQL-Datenbanken verarbeiten Daten im JSON-Format oder in einem sehr ähnlichen Format: **Couchbase** speichert JSON-Dokumente, **DynamoDB** Datensätze mit verschachtelten Attributen. **Redis** speichert als Werte zunächst nur Zeichenketten, Listen, Hashes und Mengen; echte JSON-Dokumente mit Zugriff auf einzelne Felder erst mit dem Modul **RedisJSON** (in neueren Redis-Versionen mitgeliefert). Ein JSON-Objekt sieht fast genauso aus wie ein Python-Dictionary:

```json
{
  "Name": "Alice",
  "Alter": 25,
  "Beruf": "Entwicklerin"
}
```

JSON ist besonders beliebt, weil es leichtgewichtig ist und einfach zwischen verschiedenen Programmiersprachen übertragen werden kann. Es eignet sich perfekt für Web-APIs, mobile Apps und Datenbanken, die schnell und flexibel sein müssen.
