# APIs verstehen und nutzen

APIs (Application Programming Interfaces) ermöglichen die Kommunikation zwischen verschiedenen Anwendungen und Systemen. In diesem Notebook nutzen wir das Modul `requests`, um mit Python auf freie APIs zuzugreifen, die keine Anmeldung benötigen: PokeAPI, Open-Meteo und Cat Facts.

## APIs mit Python nutzen

Mit `requests.get(url)` schickst du eine Anfrage an eine API. Die Methode `.json()` wandelt die JSON-Antwort automatisch in ein Python-Dictionary bzw. eine Liste um.

```python
import requests

antwort = requests.get("https://catfact.ninja/fact")
daten = antwort.json()
print(daten["fact"])
```

## Beispiel: Wetterdaten von Open-Meteo abfragen

Bei Open-Meteo übergibst du Breiten- und Längengrad als Parameter in der URL. Die Antwort enthält u. a. die aktuelle Temperatur.

```python
import requests

url = "https://api.open-meteo.com/v1/forecast"
parameter = {
    "latitude": 52.52,
    "longitude": 13.41,
    "current_weather": True
}

antwort = requests.get(url, params=parameter)
daten = antwort.json()
print("Aktuelle Temperatur in Berlin:", daten["current_weather"]["temperature"], "°C")
```

## Beispiel: PokeAPI

Über die PokeAPI kannst du Daten zu einzelnen Pokémon abrufen, z. B. Name, Größe und Gewicht.

```python
import requests

name = "pikachu"
antwort = requests.get(f"https://pokeapi.co/api/v2/pokemon/{name}")
daten = antwort.json()

print("Name:", daten["name"])
print("Größe:", daten["height"])
print("Gewicht:", daten["weight"])
```
