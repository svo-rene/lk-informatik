
# Lokale und Globale Variablen in Python

In Python gibt es zwei Hauptarten von Variablen: lokale Variablen und globale Variablen. Der Unterschied zwischen ihnen liegt darin, wo sie definiert sind und wo sie verwendet werden können.

### Themen:
1. Lokale Variablen
2. Globale Variablen
3. Das `global`-Schlüsselwort



## 1. Lokale Variablen

Lokale Variablen sind Variablen, die innerhalb einer Funktion definiert sind. Sie existieren nur innerhalb dieser Funktion und können von außerhalb nicht zugegriffen werden.

### Beispiel:
```python
def meine_funktion():
    x = 10  # Lokale Variable
    print("Innerhalb der Funktion:", x)

meine_funktion()
print("Außerhalb der Funktion:", x)  # Dies führt zu einem Fehler, da x lokal ist
```


````python
def meine_funktion():
    x = 10
    print("Innerhalb der Funktion:", x)

meine_funktion()
# print("Außerhalb der Funktion:", x)  # Fehler, da x lokal ist
````


## 2. Globale Variablen

Globale Variablen werden außerhalb von Funktionen definiert und können sowohl innerhalb als auch außerhalb von Funktionen verwendet werden.

### Beispiel:
```python
x = 20  # Globale Variable

def meine_funktion():
    print("Innerhalb der Funktion:", x)

meine_funktion()
print("Außerhalb der Funktion:", x)
```


````python
x = 20

def meine_funktion():
    print("Innerhalb der Funktion:", x)

meine_funktion()
print("Außerhalb der Funktion:", x)
````


## 3. Das `global`-Schlüsselwort

Das `global`-Schlüsselwort wird verwendet, um innerhalb einer Funktion auf eine globale Variable zuzugreifen und diese zu verändern. Ohne das `global`-Schlüsselwort wird innerhalb der Funktion eine neue lokale Variable erstellt.

### Beispiel:
```python
x = 20  # Globale Variable

def meine_funktion():
    global x
    x = 30  # Ändert die globale Variable
    print("Innerhalb der Funktion:", x)

meine_funktion()
print("Außerhalb der Funktion:", x)  # Der Wert von x hat sich geändert
```


````python
x = 20

def meine_funktion():
    global x
    x = 30
    print("Innerhalb der Funktion:", x)

meine_funktion()
print("Außerhalb der Funktion:", x)
````
