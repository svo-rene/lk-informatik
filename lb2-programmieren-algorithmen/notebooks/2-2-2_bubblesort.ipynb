# Einführung in Sortieralgorithmen

Sortieralgorithmen sind grundlegende Algorithmen in der Informatik, da sie häufig verwendet werden, um Daten zu organisieren und die Effizienz anderer Operationen wie Suche und Filterung zu verbessern.

### Wozu brauchen wir Sortieralgorithmen?
Sortieren ist in vielen Anwendungen wichtig, z. B.:
- Bei der Anzeige von Suchergebnissen.
- Bei der Organisation von Daten in Datenbanken.
- Um Listen zu sortieren, wie z. B. Abfahrtszeiten etc.

Es gibt verschiedene Sortieralgorithmen, die sich in ihrer Effizienz und der Art, wie sie Daten sortieren, unterscheiden. Einige Beispiele sind:
- Bubble Sort
- Insertion Sort
- Merge Sort
- Quick Sort

In dieser Lektion werden wir uns auf Bubble Sort konzentrieren, da er einfach zu verstehen ist und die Grundlagen des Sortierens vermittelt.

## Bubble Sort

Bubble Sort ist einer der einfachsten Sortieralgorithmen. Der Algorithmus funktioniert folgendermaßen:
1. Vergleiche benachbarte Elemente und tausche sie, falls sie in der falschen Reihenfolge stehen.
2. Wiederhole den Prozess, bis keine Vertauschungen mehr nötig sind.

### Funktionsweise
- In jeder Runde werden größere Elemente „nach oben gebubbelt“ (daher der Name Bubble Sort).
- Der Algorithmus durchläuft das Array mehrmals und schiebt die größeren Elemente nach rechts.
