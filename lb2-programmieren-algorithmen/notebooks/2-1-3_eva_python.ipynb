
# Eingaben und Datentypen in Python

In diesem Notebook werden wir uns ansehen, wie Benutzereingaben in Python verarbeitet werden und wie verschiedene Datentypen verwendet werden. Datentypen bestimmen, welche Operationen auf einer Variable durchgeführt werden können.

### Themen:
1. Eingaben vom Benutzer
2. Wichtige Datentypen (int, float, str, bool)
3. Typumwandlung (Casting)



## 1. Eingaben vom Benutzer

In Python wird die Funktion `input()` verwendet, um Eingaben vom Benutzer zu erhalten. Die Eingaben werden standardmäßig als Zeichenkette (`str`) gespeichert.

### Beispiel:
```python
name = input("Wie heißt du? ")
print("Hallo, " + name + "!")
```


````python
name = input("Wie heißt du? ")
print("Hallo, " + name + "!")
````


## 2. Wichtige Datentypen

Python unterstützt verschiedene Datentypen. Hier sind die wichtigsten:
- `int`: Ganze Zahlen
- `float`: Gleitkommazahlen
- `str`: Zeichenketten (Text)
- `bool`: Wahrheitswerte (True, False)

### Beispiele:
```python
ganze_zahl = 10  # int
kommazahl = 3.14  # float
text = "Hallo"  # str
wahrheitswert = True  # bool
```


````python
ganze_zahl = 10
kommazahl = 3.14
text = "Hallo"
wahrheitswert = True
print(ganze_zahl, kommazahl, text, wahrheitswert)
````


## 3. Typumwandlung (Casting)

Manchmal ist es notwendig, den Typ einer Variablen zu ändern, zum Beispiel, wenn man eine Benutzereingabe (immer `str`) in eine Zahl umwandeln möchte. Dafür gibt es Funktionen wie `int()`, `float()`, und `str()`.

### Beispiel:
```python
alter = input("Wie alt bist du? ")  # Eingabe als Zeichenkette
alter = int(alter)  # Umwandlung in eine ganze Zahl
print("In 10 Jahren wirst du", alter + 10, "Jahre alt sein.")
```


````python
alter = input("Wie alt bist du? ")
alter = int(alter)
print("In 10 Jahren wirst du", alter + 10, "Jahre alt sein.")
````
