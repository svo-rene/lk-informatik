
# Unterprogramme in Python

Unterprogramme sind wiederverwendbare Codeblöcke, die unter einem Namen zusammengefasst sind. Sie ermöglichen es, den Code zu strukturieren und zu modularisieren, indem bestimmte Aufgaben in Unterprogramme ausgelagert werden.

## 1. Definition von Funktionen

Eine Funktion wird mit dem Schlüsselwort `def` definiert. Der Funktionsname wird gefolgt von Klammern `()` und einem Doppelpunkt `:`. Der Funktionscode wird eingerückt.

### Beispiel:
```python
def begrüßung():
    print("Hallo!")
```

Um diese Funktion auszuführen, muss sie mit ihrem Namen aufgerufen werden:
```python
begrüßung()
```


````python
def begrüßung():
    print("Hallo!")

begrüßung()
````


## 2. Funktionsparameter und Rückgabewerte

Funktionen können Parameter entgegennehmen, die beim Aufruf übergeben werden, und sie können auch Werte zurückgeben.

### Beispiel:
```python
def begrüßung(name):
    print("Hallo, " + name + "!")
    
begrüßung("Anna")  # Übergabe des Namens "Anna"
```

Funktionen können auch einen Wert mit dem Schlüsselwort `return` zurückgeben:
```python
def quadriere(x):
    return x * x

ergebnis = quadriere(5)
print(ergebnis)  # Gibt 25 aus
```


````python
def begrüßung(name):
    print("Hallo, " + name + "!")

begrüßung("Anna")
begrüßung("Bob")
````

````python
def quadriere(x):
    return x * x

ergebnis = quadriere(5)
print(ergebnis)
````

Funktionen können mehrere Parameter haben, um flexibler zu sein. Parameter können auch Standardwerte haben.

````python
def begrüße(vorname, nachname, titel="Dr."):
    if titel:
        return "Hallo " + titel + " " + vorname + " " + nachname + "!"
    else:
        return "Hallo " + vorname + " " + nachname + "!"

print(begrüße("Max", "Mustermann"))
print(begrüße("Maria", "Musterfrau", "Prof."))
````


## 3. Unterschiede Funktionen und Prozeduren

In der Programmierung wird oft zwischen Prozeduren und Funktionen unterschieden, obwohl Python diesen Unterschied syntaktisch nicht macht.


````python
# Prozedur (kein Rückgabewert)
def grüße_nutzer(name):
    print("Willkommen,", name)

# Funktion (gibt einen Wert zurück)
def berechne_summe(a, b):
    return a + b

# Verwendung
grüße_nutzer("Alice")  # Prozedur
ergebnis = berechne_summe(3, 4)  # Funktion
print("Die Summe ist:", ergebnis)
````
