# Suchalgorithmen: Lineare und Binäre Suche

## Lineare/Sequenzielle Suche
Die lineare Suche ist der einfachste Suchalgorithmus. Hierbei werden die Elemente einer Liste nacheinander durchsucht, 
bis das gewünschte Element gefunden wird oder das Ende der Liste erreicht ist. 
Die lineare Suche hat eine Zeitkomplexität von **???**.

### Funktionsweise:
1. Beginne beim ersten Element der Liste.
2. Vergleiche das aktuelle Element mit dem Suchwert.
3. Wenn der Suchwert gefunden wurde, stoppe die Suche.
4. Wenn das Ende der Liste erreicht ist, ohne den Suchwert zu finden, ist das Element nicht vorhanden.

## Binäre Suche
Die binäre Suche ist ein effizienterer Algorithmus, der jedoch voraussetzt, dass die Liste sortiert ist. 
Er teilt die Liste bei jedem Schritt in zwei Hälften und sucht nur in der Hälfte weiter, die das gesuchte Element enthalten könnte. 
Die binäre Suche hat eine Zeitkomplexität von **???**.

### Funktionsweise:
1. Finde das mittlere Element der Liste.
2. Vergleiche das mittlere Element mit dem Suchwert:
   - Wenn sie gleich sind, ist die Suche abgeschlossen.
   - Wenn der Suchwert kleiner ist, suche in der linken Hälfte weiter.
   - Wenn der Suchwert größer ist, suche in der rechten Hälfte weiter.
3. Wiederhole den Vorgang, bis das Element gefunden wird oder die Suchregion leer ist.
