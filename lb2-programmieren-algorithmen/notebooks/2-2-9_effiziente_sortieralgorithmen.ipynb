## 1. Arbeiten mit Modulen

### Theorie: Module in Python

Ein Modul in Python ist eine Datei, die Python-Code (Funktionen, Klassen, Variablen) enthält und in anderen Programmen importiert werden kann. Der Zweck von Modulen ist:
- Code-Wiederverwendbarkeit: Funktionen oder Klassen müssen nicht mehrfach geschrieben werden.
- Struktur: Programme können logisch in kleinere, wartbare Teile zerlegt werden.

**Erstellen eines Moduls**  
Ein Modul ist eine `.py`-Datei, die Python-Code enthält. Beispiel:

```python
# Datei: mein_modul.py
def begruessung(name):
    return "Hallo, " + name + "!"
```

**Importieren eines Moduls**  
Module können mit `import` oder `from ... import` in einem Python-Programm verwendet werden.

```python
# Datei: main.py
import mein_modul

print(mein_modul.begruessung("Anna"))
```

**Vorteile von Modulen**
- Reduzierte Code-Duplikation.
- Bessere Lesbarkeit und Wartbarkeit.
- Standardbibliotheken und Drittanbieter-Module sind sofort verfügbar.

# 3. Erklärung: Datensätze als Listen

Ein Datensatz besteht aus mehreren Werten, die zusammengehören, z. B. Name, Vorname, Straße, PLZ und Ort einer Person. In Python speichern wir einen Datensatz als **Liste**. Alle Datensätze zusammen bilden wieder eine Liste – eine **Liste von Listen**.

**Beispiele:**

```python
# Ein Datensatz mit 5 Werten
person = ["Krause", "Stefanie", "Brandenburgische Str. 20", "74343", "Sachsenheim"]

# Zugriff auf einzelne Werte
print(person[0])  # Ausgabe: Krause
print(person[3])  # Ausgabe: 74343

# Eine Liste von Datensätzen
daten = [person, ["Schmitz", "Jan", "Hauptstr. 1", "01099", "Dresden"]]
print(daten[1][0])  # Ausgabe: Schmitz (Name im zweiten Datensatz)
```

Die Reihenfolge der Werte ist in jedem Datensatz gleich. Deshalb steht der Name immer an Index 0 und die PLZ immer an Index 3.


## 4. Arbeiten mit CSV-Dateien

Das CSV-Format (Comma-Separated Values) ist ein weit verbreitetes Dateiformat zur Speicherung tabellarischer Daten. Jede Zeile entspricht einer Datenzeile, Spalten sind durch ein Trennzeichen (z. B. Komma oder Semikolon) getrennt.

**Lesen von CSV-Dateien**  
CSV-Dateien können mit in ein Progamm  eingelesen werden:

```python
def textAusDatei(dateiname):

    # lädt einen Text aus einer Datei und gibt sie als
    # Zeichenkette zurück
    # beachte: setzt die Codierung iso-8859-1 voraus
    
    datei = open(dateiname, 'r', encoding='iso-8859-1')
    text = datei.read()
    datei.close()
    return text
```

**Schreiben von CSV-Dateien**  
Daten können in CSV-Dateien geschrieben werden:

```python
def textInDateiSpeichern(dateiname, text):

    # speichert eine Zeichenkette in einer Datei
    # beachte: benutzt die Codierung iso-8859-1
    
    datei = open(dateiname, 'w', encoding='iso-8859-1')
    datei.write(text)
    datei.close()

```

**Häufige Anwendungsfälle**
- Datenimport aus Tabellen oder Datenbanken.
- Speicherung von Ergebnissen, die von Python-Skripten erzeugt wurden.
- Analyse und Visualisierung von Daten.

**Vorteile von CSV**
- Einfach zu erstellen und zu lesen.
- Unterstützt von vielen Tools (z. B. Excel, Datenbanken).

### Aufgabe 2: Sortieren von Datensätzen

Bevor der gegebene Sortieralgorithmus auf diesen Daten operieren kann, müssen die Daten zunächst in eine Listenform überführt werden.

Diese Datentransformationen können mit Funktionen des Python-Programms listenoperationen_zum_laden_und_speichern.py durchgeführt werden:

Wenn man jetzt Quicksort anwenden möchte, muss man zum einen auf die vorgegebenen Funktionen zur Datentransformation zurückgreifen. Zum anderen muss man die Vergleichsoperation zum Vergleich der Datensätze anpassen. Der folgende Quelltext zeigt, wie das geht.

1) Erläutere, was gegenüber einem Sortierprogramm für eine Liste von Zahlen verändert wurde (Vergleich der Datensätze, Laden und Speichern).

Hinweis: Diese Quicksort-Fassung teilt anders als die im Kurs. Das Pivot ist das erste Element, `links` und `rechts` laufen aufeinander zu. Auf schon sortierte Daten angewandt bricht sie mit `RecursionError` ab.

````python
from listenoperationen_zum_laden_und_speichern import *

# Vergleich von zwei Datensätzen (nach dem Namen an Index 0)
def kleiner(d1, d2):
    return d1[0] < d2[0]

# Sortieralgorithmus
def quicksort(L, anfang, ende):
    pivot = L[anfang]
    links = anfang
    rechts = ende
    while links <= rechts:
        while kleiner(L[links], pivot):
            links = links+1
        while kleiner(pivot, L[rechts]):
            rechts = rechts-1
        if links <= rechts:
            if links < rechts:
                h = L[links]
                L[links] = L[rechts]
                L[rechts] = h
            links = links+1
            rechts = rechts-1
    if anfang < rechts:
        L = quicksort(L, anfang, rechts)
    if links < ende:
        L = quicksort(L, links, ende)
    return L

# Test
text = textAusDatei('adressdaten.csv')
listeDaten = datensaetzeAusText(text)
listeDatenSortiert = quicksort(listeDaten, 0, len(listeDaten)-1)
textNeu = textAusDatensaetzen(listeDatenSortiert)
textInDateiSpeichern('adressdaten_sortiert_name.csv', textNeu)
````

### Aufgabe 5: Weitere Sortieralgorithmen anpassen

Gegeben ist ein Programm zum Sortieralgorithmus Selectionsort. Passe es so an, dass mit dem veränderten Programm die oben gegebenen Adressdaten nach dem Nachnamen / der Postleitzahl sortiert werden.

````python
# Sortieralgorithmus
def selectionsort(L):
    n = len(L)
    i = 0
    while i < n-1:
        minpos = i
        m = i+1
        while m < n:
            if L[m] < L[minpos]:
                minpos = m
            m = m+1
        h = L[i]
        L[i] = L[minpos]
        L[minpos] = h
        i = i+1
    return L

# Test
L = [25, 17, 32, 56, 25, 19, 8, 66, 29, 6, 20, 29]
print(L)
L = selectionsort(L)
print(L)
````
