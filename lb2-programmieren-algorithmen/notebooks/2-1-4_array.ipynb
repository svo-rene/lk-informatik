Listen sind eine der grundlegenden Datenstrukturen in Python. Sie ermöglichen es, mehrere Elemente in einer geordneten Folge zu speichern. Listen sind veränderbar (mutierbar), das heißt, ihre Elemente können nach der Erstellung geändert werden.

## 1. Erstellen von Listen

Eine Liste kann mit eckigen Klammern `[]` erstellt werden, und ihre Elemente werden durch Kommas getrennt.

### Beispiel:
```python
meine_liste = [1, 2, 3, 4, 5]
print(meine_liste)
```


````python
# Eine Liste mit Zahlen
zahlen = [1, 2, 3, 4, 5]
print("Zahlenliste:", zahlen)

# Eine Liste mit Strings
früchte = ["Apfel", "Banane", "Kirsche"]
print("Früchteliste:", früchte)

# Eine gemischte Liste
gemischt = [1, "Zwei", 3.0, True]
print("Gemischte Liste:", gemischt)
````


## 2. Zugriff auf Listenelemente

Auf die Elemente einer Liste kann mit Hilfe von Indizes zugegriffen werden. Python-Listen sind nullbasiert, das heißt, das erste Element hat den Index `0`.

### Beispiel:
```python
meine_liste = ["Apfel", "Banane", "Kirsche"]
print(meine_liste[0])  # Gibt "Apfel" aus
print(meine_liste[-1])  # Gibt "Kirsche" aus
```


````python
meine_liste = ["Apfel", "Banane", "Kirsche"]
print(meine_liste[0])
print(meine_liste[-1])
````


## 3. Listen bearbeiten (Hinzufügen, Entfernen, Ändern)

Python bietet eine Vielzahl an Funktionen und Methoden zur Arbeit mit Listen, z.B. `len()`, `append()`, `remove()`, `sort()`, usw. Elemente können zu einer Liste hinzugefügt, aus ihr entfernt oder geändert werden.

### Beispiel:
```python
zahlen = [1, 2, 3, 4, 5]
zahlen[2] = 30  # Ändert das dritte Element
print(zahlen)

print(len(zahlen)) # Gibt die Anzahl der Elemente der Liste wieder

zahlen.append(6)  # Fügt eine 6 hinzu
print(zahlen)

zahlen.remove(30)  # Entfernt die 30
print(zahlen)
```


````python
zahlen = [1, 2, 3, 4, 5]
zahlen[2] = 30
print(zahlen)
print(len(zahlen))
zahlen.append(6)
print(zahlen)
zahlen.remove(30)
print(zahlen)
````


## 4. Listen mit Schleifen ausgeben

Eine `for`-Schleife kann verwendet werden, um über eine Liste zu iterieren.

### Beispiel:
```python
zahlen = [1, 2, 3, 4, 5]
for zahl in zahlen:
    print(zahl)
```


````python
# For-Schleife 1
zahlen = [1, 2, 3, 4, 5]
for zahl in zahlen:
    print(zahl)

# For-Schleife 2
for i in range (len(zahlen),0,-1):
    print(zahlen[i-1])

# While-Schleife
i = 0
while i < len(zahlen):
    zahlen[i] = zahlen[i] * 2
    i = i + 1
print(zahlen)
````
