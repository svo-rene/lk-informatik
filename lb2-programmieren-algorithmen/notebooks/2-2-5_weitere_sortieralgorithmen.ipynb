## Insertionsort

Der Insertionsort-Algorithmus ist im folgenden Python-Quellcode gegeben. Beschreiben Sie anschließend den Algorithmus als Pseudocode.

````python
# Python-Implementierung von Insertionsort
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j = j - 1
        arr[j + 1] = key
    return arr
````

## Selectionsort

Der Selectionsort-Algorithmus funktioniert, indem er wiederholt das kleinste Element im unsortierten Teil der Liste findet und dieses an den Anfang verschiebt.

### Pseudocode für Selectionsort

```
Selectionsort(A):
Eine Schleife von i bis zur Länge der Liste minus 2:
        Die Variable minIndex wird auf i gesetzt.
        Eine innere Schleife von j = i + 1 bis zur Länge der Liste - 1:
            Wenn das Element an Position j kleiner ist als das Element an Position minIndex:
                Setze minIndex auf j.
        Tausche das Element an Position i mit dem Element an Position minIndex.
```
