## Weitere Funktionen zum Arbeiten mit Texten

**Merke:** Wichtige Methoden zum Arbeiten mit Text:

- `startswith('Text')`: Prüft, ob ein Textanfang übereinstimmt.
- `endswith('Text')`: Prüft, ob ein Textende übereinstimmt.
- `'Teiltext' in Text`: Prüft, ob Teiltext enthalten ist.
- `isdigit()`: Prüft, ob ein Zeichen eine Zahl ist.
- `lower()`: Wandelt Text in Kleinbuchstaben um.
- `upper()`: Wandelt Text in Großbuchstaben um.
- `strip()`: Entfernt Leerzeichen am Anfang und Ende.
- `split()`: Zerlegt einen Text in Teile (z. B. Vorname, Nachname).

````python
# Aufgabe 1
import csv

with open("sportveranstaltung.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    max_zeit = 0
    name = ""
    for row in reader:
        zeit = int(row["Zeit"])
        if zeit > max_zeit:
            max_zeit = zeit
            name = row["Name"]
    print(name)
````

````python
# Aufgabe 3
import csv

with open("sportveranstaltung.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    staedte = {}

    for row in reader:
        stadt = row["Stadt"]
        name = row["Name"]
        if stadt not in staedte:
            staedte[stadt] = []
        staedte[stadt].append(name)

    for stadt in staedte:
        print(stadt + ":", ", ".join(staedte[stadt]))
````

````python
# Aufgabe 4
import csv

with open("sportveranstaltung.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    zaehler = {}

    for row in reader:
        stadt = row["Stadt"]
        if stadt not in zaehler:
            zaehler[stadt] = 0
        zaehler[stadt] += 1

    for stadt in zaehler:
        print(stadt + ":", zaehler[stadt])
````

````python
# Aufgabe 7
import csv

namen = []

with open("sportveranstaltung.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    for row in reader:
        name = row["Name"]
        teile = name.split()  # zerlegt z. B. "Anna Müller" in ["Anna", "Müller"]
        nachname = teile[-1]  # nimmt das letzte Wort als Nachname
        namen.append((nachname, name))  # speichert Nachname + Originalname

namen.sort()  # sortiert nach Nachname (erstes Element im Tupel)

print(namen)
````

````python
# Aufgabe 12
# isdigit() prüft nur, ob ein ganzer Text nur aus Ziffern besteht.
# Um zu prüfen, ob irgendwo im Namen eine Ziffer vorkommt, gehen wir
# mit einer Schleife Zeichen für Zeichen durch den Text.
import csv

def enthaelt_ziffer(text):
    for zeichen in text:
        if zeichen.isdigit():
            return True
    return False

with open("sportveranstaltung.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    for row in reader:
        if enthaelt_ziffer(row['Name']):
            print(row['Name'])
````
