# JSON: Ein aktuelles Standardformat

JSON (JavaScript Object Notation) ist ein leichtgewichtiges, textbasiertes Datenformat, das von fast allen modernen Programmiersprachen unterstützt wird und zu einem Standard für Web-APIs geworden ist.

In diesem Notebook nutzen wir das Modul `json`, um JSON-Text und JSON-Dateien in Python-Datenstrukturen umzuwandeln. Wie man JSON mit `requests` direkt von einer API abruft, lernst du auf Seite 3.11.

## JSON in Python verarbeiten

Mit `json.loads()` wandelst du einen JSON-Text in ein Dictionary (bzw. eine Liste) um. Ein verschachteltes JSON-Objekt wird dabei zu verschachtelten Dictionaries und Listen, die du wie gewohnt mit eckigen Klammern ansprichst.

```python
import json

text = '{"name": "Max Mustermann", "age": 30, "skills": ["JavaScript", "Python", "JSON"]}'

daten = json.loads(text)  # wandelt den JSON-Text in ein Dictionary um
print(daten["name"])       # Max Mustermann
print(daten["skills"][1])  # Python

# JSON-Datei einlesen: json.load() (ohne "s") mit einer geöffneten Datei
with open("wetter_dresden.json", encoding="utf-8") as datei:
    wetter = json.load(datei)
print(wetter["current_weather"]["temperature"], "°C")   # 14.8 °C
```

```python
import json

with open("wetter_dresden.json", encoding="utf-8") as datei:
    wetter = json.load(datei)

# 1) Windgeschwindigkeit mit Einheit
print("Wind:", wetter["current_weather"]["windspeed"],
      wetter["current_weather_units"]["windspeed"])        # Wind: 11.2 km/h

# 2) Niedrigste stündliche Temperatur mit Uhrzeit
temps = wetter["hourly"]["temperature_2m"]
zeiten = wetter["hourly"]["time"]
minimum = min(temps)
position = temps.index(minimum)
print("Minimum:", minimum, "°C um", zeiten[position][11:])  # Minimum: 11.9 °C um 08:00
```
