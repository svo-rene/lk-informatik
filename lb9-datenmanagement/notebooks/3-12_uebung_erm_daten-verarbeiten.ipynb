**Musterlösung a** (mit `csv.reader`)

```python
import csv

temperaturen = []
with open("wetterdaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.reader(csvfile, delimiter=";")
    next(reader)
    for row in reader:
        if row[1] == "Dresden":
            temperaturen.append(row[2])
for wert in temperaturen:
    print(wert)
```

**Musterlösung a** (mit `csv.DictReader`)

```python
import csv

temperaturen = []
with open("wetterdaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    for row in reader:
        if row["Ort"] == "Dresden":
            temperaturen.append(row["Temperatur"])
for wert in temperaturen:
    print(wert)
```

**Musterlösung b** (mit `csv.reader`)

```python
import csv

werte = []
with open("wetterdaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.reader(csvfile, delimiter=";")
    next(reader)
    for row in reader:
        temperatur = int(row[2])
        datum = row[0]
        ort = row[1]
        werte.append((temperatur, datum, ort))
max_eintrag = max(werte)
print("Höchste Temperatur:", max_eintrag[0], "am", max_eintrag[1], "in", max_eintrag[2])
```

**Musterlösung b** (mit `csv.DictReader`)

```python
import csv

werte = []
with open("wetterdaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    for row in reader:
        temperatur = int(row["Temperatur"])
        datum = row["Datum"]
        ort = row["Ort"]
        werte.append((temperatur, datum, ort))
max_eintrag = max(werte)
print("Höchste Temperatur:", max_eintrag[0], "am", max_eintrag[1], "in", max_eintrag[2])
```

**Musterlösung c** (mit `csv.reader`)

```python
import csv

orte = []
zaehlungen = []
with open("wetterdaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.reader(csvfile, delimiter=";")
    next(reader)
    for row in reader:
        ort = row[1]
        if ort in orte:
            index = orte.index(ort)
            zaehlungen[index] += 1
        else:
            orte.append(ort)
            zaehlungen.append(1)
for i in range(len(orte)):
    print(orte[i] + ":", zaehlungen[i])
```

**Musterlösung c** (mit `csv.DictReader`)

```python
import csv

zaehlungen = {}
with open("wetterdaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    for row in reader:
        ort = row["Ort"]
        if ort not in zaehlungen:
            zaehlungen[ort] = 0
        zaehlungen[ort] += 1
for ort in zaehlungen:
    print(ort + ":", zaehlungen[ort])
```

**Musterlösung d** (mit `csv.reader`)

```python
import csv

werte = []
with open("wetterdaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.reader(csvfile, delimiter=";")
    next(reader)
    for row in reader:
        if row[1] == "Dresden":
            werte.append(int(row[3]))
durchschnitt = sum(werte) / len(werte)
print("Durchschnitt Luftfeuchtigkeit in Dresden:", durchschnitt)
```

**Musterlösung d** (mit `csv.DictReader`)

```python
import csv

werte = []
with open("wetterdaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    for row in reader:
        if row["Ort"] == "Dresden":
            werte.append(int(row["Luftfeuchtigkeit"]))
durchschnitt = sum(werte) / len(werte)
print("Durchschnitt Luftfeuchtigkeit in Dresden:", durchschnitt)
```
