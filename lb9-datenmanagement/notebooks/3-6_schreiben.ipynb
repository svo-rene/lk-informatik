# Daten in CSV-Dateien schreiben

Das Speichern von Daten im CSV-Format ist eine zentrale Methode, um strukturierte Informationen systemübergreifend weiterzugeben, zu archivieren oder automatisiert auszuwerten.

## Daten mit csv.writer() als csv-Datei speichern

Das folgende Beispiel zeigt, wie mehrere strukturierte Datensätze mithilfe des Moduls `csv` in eine neue CSV-Datei geschrieben werden. Die Daten werden als Liste von Zeilen definiert und anschließend vollständig exportiert.

```python
import csv

daten = [
    ["Name", "Alter", "Note"],   # Kopfzeile (Spaltenüberschriften)
    ["Anna", "17", "1.3"],       # Erste Datenzeile
    ["Ben", "16", "2.0"]         # Zweite Datenzeile
]

with open("ausgabe.csv", "w", newline="", encoding="utf-8") as datei:
    writer = csv.writer(datei, delimiter=";")  # Erzeuge ein CSV-Schreibobjekt mit Semikolon als Trennzeichen
    writer.writerows(daten)  # Schreibe alle Datenzeilen auf einmal in die Datei
```

> **Hinweis: Warum `newline=""` und `utf-8`?** Das `csv`-Modul schreibt Zeilenenden selbst als `\r\n`; ohne `newline=""` würde Python unter Windows daraus `\r\r\n` machen – dann erscheint nach jedem Datensatz eine Leerzeile. `utf-8` sichert die Darstellung von Umlauten und Sonderzeichen.

## Daten mit csv.DictWriter() als csv-Datei schreiben

In diesem Beispiel werden strukturierte Datensätze in Form von Dictionaries in eine CSV-Datei geschrieben. Durch die Verwendung von Spaltennamen können die Daten gezielt und übersichtlich gespeichert werden.

```python
import csv

daten = [
    {"Name": "Anna", "Alter": "17", "Note": "1.3"},
    {"Name": "Ben", "Alter": "16", "Note": "2.0"}
]

with open("ausgabe_dict.csv", "w", newline="", encoding="utf-8") as datei:
    feldnamen = ["Name", "Alter", "Note"]  # Definiere die Reihenfolge der Spalten

    writer = csv.DictWriter(datei, fieldnames=feldnamen, delimiter=";")  # Erzeuge ein CSV-Schreibobjekt mit Spaltennamen
    writer.writeheader()  # Schreibe zuerst die Kopfzeile
    writer.writerows(daten)  # Schreibe alle Dictionaries in die Datei
```

> **Hinweis: Dateien schreiben, bestehende Dateien ändern**
>
> Der Modus `"r"` steht für *read* – Datei wird nur gelesen.
>
> Der Modus `"w"` steht für *write* – Datei wird neu geschrieben (bestehende Inhalte werden gelöscht).
>
> Der Modus `"a"` steht für *append* – neue Inhalte werden am Ende der Datei angehängt.

## Einzelne Zeilen gezielt schreiben

Neben dem Schreiben ganzer Tabellen kann es sinnvoll sein, einzelne Ergebnisse oder aggregierte Daten schrittweise in eine CSV-Datei zu schreiben – etwa zur Dokumentation von Berechnungen oder zur Protokollierung.

### Beispiel: Einzelwerte in CSV speichern

```python
import csv

# Beispielwerte einer Temperaturauswertung
anzahl_messungen = 120
durchschnitt = 21.7
maximum = 34.5
minimum = 15.2

with open("auswertung_temperatur.csv", mode="w", encoding="utf-8", newline="") as file:
    writer = csv.writer(file, delimiter=";")

    writer.writerow(["Anzahl Messungen", anzahl_messungen])
    writer.writerow(["Durchschnittstemperatur (°C)", durchschnitt])

    writer.writerow([])  # Leerzeile zur Trennung
    writer.writerow(["Maximale Temperatur (°C)", maximum])
    writer.writerow(["Minimale Temperatur (°C)", minimum])
```

### Beispiel: Dictionary in CSV speichern

```python
import csv

statistik = {16: 3, 17: 5, 18: 2}

with open("schueler_nach_alter.csv", "w", newline="", encoding="utf-8") as datei:
    writer = csv.writer(datei, delimiter=";")
    writer.writerow(["Alter", "Anzahl"])

    for alter, anzahl in statistik.items():
        writer.writerow([alter, anzahl])
```

**Merke:** `writerow()` schreibt genau eine einzelne Zeile in die CSV-Datei. `writerows()` schreibt mehrere Zeilen auf einmal. Nutze `writerow()` bei Schleifen oder Einzelergebnissen, `writerows()` bei kompletten Datensätzen.

```python
# Komplexaufgabe
import csv

MIN_TEMP, MAX_TEMP = -30, 50   # dieselben Grenzen wie auf Seite 3.5

# Schritt 1: Einlesen
sensordaten = []
with open("sensordaten.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=";")
    for row in reader:
        sensordaten.append(row)

# Schritt 2: Datensätze mit ungültigem Code entfernen
gueltige_daten = [e for e in sensordaten if e["Code"].isdigit()]

# Schritt 3: Mittelwert NUR aus den realistischen Temperaturen berechnen
realistische = []
for e in gueltige_daten:
    t = float(e["Temperatur"])
    if MIN_TEMP <= t <= MAX_TEMP:
        realistische.append(t)
mittelwert = round(sum(realistische) / len(realistische), 1)

# Schritt 4: Ausreißer durch diesen Mittelwert ersetzen
for e in gueltige_daten:
    t = float(e["Temperatur"])
    e["Temperatur"] = t if MIN_TEMP <= t <= MAX_TEMP else mittelwert

# Schritt 5: Sensortyp aus SensorInfo herauslösen ("G18_Typ-A" -> "Typ-A")
for e in gueltige_daten:
    e["Typ"] = e["SensorInfo"].split("_")[1]

# Schritt 6: Bereinigte Daten speichern
feldnamen = ["Zeit", "Code", "Temperatur", "SensorInfo", "Typ"]
with open("sensordaten_bereinigt.csv", "w", newline="", encoding="utf-8") as datei:
    writer = csv.DictWriter(datei, fieldnames=feldnamen, delimiter=";")
    writer.writeheader()
    writer.writerows(gueltige_daten)

# Schritt 7: Datensätze über 30 °C separat speichern
hohe_temperaturen = [e for e in gueltige_daten if e["Temperatur"] > 30]
with open("temperatur_hoch.csv", "w", newline="", encoding="utf-8") as datei:
    writer = csv.DictWriter(datei, fieldnames=feldnamen, delimiter=";")
    writer.writeheader()
    writer.writerows(hohe_temperaturen)

print("Mittelwert der realistischen Werte:", mittelwert)
print(len(gueltige_daten), "gültige Datensätze,", len(hohe_temperaturen), "davon über 30 °C")
```

```python
# Vorabitur
import csv

def ist_gueltiger_code(code):
    if len(code) != 3:
        return False
    if code[0] not in ('G', 'S'):
        return False
    if not code[1].isdigit() or code[1] == '0':
        return False
    if not code[2].isdigit():
        return False
    return True

gesamt = 0
getraenke = 0
snacks = 0
zaehler = {}

with open("snackbot.csv", newline="", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile, delimiter=",")
    for row in reader:
        code = row["Produktcode"]
        if ist_gueltiger_code(code):
            gesamt += 1
            if code[0] == "G":
                getraenke += 1
            else:
                snacks += 1
            if code not in zaehler:
                zaehler[code] = 0
            zaehler[code] += 1

# Meistverkauftes Produkt finden
bestseller = ""
bestseller_anzahl = 0
for code in zaehler:
    if zaehler[code] > bestseller_anzahl:
        bestseller = code
        bestseller_anzahl = zaehler[code]

with open("auswertung_verkauf_automat.csv", "w", newline="", encoding="utf-8") as datei:
    writer = csv.writer(datei, delimiter=";")
    writer.writerow(["Gesamtverkäufe", gesamt])
    writer.writerow(["Anzahl Getränke", getraenke])
    writer.writerow(["Anzahl Snacks", snacks])
    writer.writerow(["Meistverkauftes Produkt", bestseller])
    writer.writerow(["Anzahl Verkäufe Bestseller", bestseller_anzahl])
```
