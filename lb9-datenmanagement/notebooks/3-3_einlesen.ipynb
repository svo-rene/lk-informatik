## CSV-Dateien als Listen einlesen

Eine CSV-Datei kann zeilenweise eingelesen und jede Zeile in eine Liste umgewandelt werden:

- `open()` öffnet die Datei zum Lesen.
- `csv.reader()` wandelt jede Zeile in eine Liste um.
- `for row in reader` geht Zeile für Zeile durch die Datei.

```python
import csv

with open('schueler.csv', newline='', encoding='utf-8') as csvfile:
    reader = csv.reader(csvfile, delimiter=';')
    for row in reader:
        print(row)
```

Jede Zeile wird zu einer Liste von Texten (Strings).

## CSV-Dateien als Dictionaries einlesen

Wenn die CSV-Datei eine Kopfzeile mit Spaltennamen hat, können wir die Daten direkt als Wörterbücher laden:

- `csv.DictReader()` erstellt für jede Zeile ein Dictionary, wobei die Spaltennamen als Schlüssel verwendet werden.
- Jede Ausgabe ist ein Wörterbuch, für `schueler.csv` z. B. `{'Name': 'Anna', 'Alter': '17', 'Note': '1.3'}`.

```python
import csv

with open('produkte.csv', newline='', encoding='utf-8') as csvfile:
    reader = csv.DictReader(csvfile, delimiter=';')
    for row in reader:
        print(row)
```

Jede Zeile wird zu einem Wörterbuch (`dict`), wobei die Spaltennamen als Schlüssel verwendet werden.

## Wichtige Parameter beim Einlesen von CSV-Dateien

- **`newline=''`**: Schaltet die automatische Umwandlung von Zeilenenden durch `open()` ab und überlässt sie dem `csv`-Modul. Nur so werden Felder, die in Anführungszeichen einen Zeilenumbruch enthalten, richtig gelesen; beim Schreiben verhindert es unter Windows zusätzliche Leerzeilen.
- **`encoding='utf-8'`**: Sorgt für korrekte Darstellung von Umlauten und Sonderzeichen. Hat die Datei ein unsichtbares BOM am Anfang (typisch für „CSV UTF-8“ aus Excel/LibreOffice), landet es sonst im ersten Spaltennamen (`'\ufeffName'`) und `row['Name']` meldet einen `KeyError` – dann hilft `encoding='utf-8-sig'`. Bei Dateien älterer Windows-Programme hilft oft `encoding='cp1252'`.
- **`delimiter`**: Gibt an, welches Zeichen die Werte trennt. In Deutschland häufig das Semikolon (`;`), manchmal auch das Komma (`,`).

> **Hinweis:** Beim Einlesen von CSV-Dateien ist das richtige Trennzeichen (`delimiter`) sehr wichtig! In Deutschland wird oft das Semikolon (`;`) statt dem Komma (`,`) verwendet, weil wir das Komma häufig zur Trennung von Dezimalzahlen benutzen.
