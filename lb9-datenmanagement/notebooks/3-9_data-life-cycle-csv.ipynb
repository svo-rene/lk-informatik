## 1. Eigene Umfrage erstellen

### Aufgabe 1: Denk dir zwei Umfragefragen aus

Überlege dir zwei Fragen:

- **Frage 1:** mit einer Antwort als Wort (z. B. Musikrichtung, Lieblingstier, Lieblingsspiel, Lieblingsverein, Lieblingsfarbe)
- **Frage 2:** mit einer Antwort als Zahl, auch Dezimalzahlen sind erlaubt (z. B. Schuhgröße, Anzahl Haustiere, Schlafdauer in Stunden wie 7.5)

Erstelle in LibreOffice Calc eine Tabelle mit mindestens **10 Datensätzen** (Spalten: Nr, Antwort_Wort, Antwort_Zahl). Trage statt Namen nur eine laufende Nummer ein – so bleibt die Umfrage anonym.

Baue absichtlich **2–3 Fehler** ein, z. B.:

- eine leere Zelle
- Antworten vertauscht (Zahl bei Wort, Wort bei Zahl)
- Buchstaben in der Zahlenspalte

Speichere deine Datei als `umfrage.csv` (UTF-8, Semikolon als Trennzeichen, Dezimalzahlen mit Punkt) im selben Ordner wie dieses Notebook.

## 2. Daten mit Python einlesen

```python
import csv

daten = []
with open('umfrage.csv', newline='', encoding='utf-8') as file:
    reader = csv.reader(file, delimiter=';')
    next(reader)  # Kopfzeile überspringen
    for zeile in reader:
        daten.append(zeile)

for eintrag in daten:
    print(eintrag)
```

### Verständnis: Wie funktioniert das Einlesen?

Beschreibe in eigenen Worten:

- Was genau passiert in der Schleife `for zeile in reader`?
- Was speichert die Liste `daten`?
- Wie sehen die einzelnen Elemente darin aus?

Tipp: Gib `print(daten)` oder `print(daten[0])` aus und analysiere, was darin steht.

```python
# Zum Ausprobieren:
print(daten[0])
```

## 3. Daten bereinigen

```python
def ist_zahl(text):
    """True, wenn text eine ganze Zahl oder Dezimalzahl ist (z. B. '3', '7.5', '7,5')."""
    try:
        float(text.replace(',', '.'))
        return True
    except ValueError:
        return False

bereinigt = []

for zeile in daten:
    if len(zeile) == 3:
        nr = zeile[0].strip()
        wort = zeile[1].strip()
        zahl = zeile[2].strip().replace(',', '.')   # Dezimalkomma in Punkt umwandeln

        if wort and not ist_zahl(wort) and ist_zahl(zahl):
            bereinigt.append([nr, wort, zahl])
```

### Verständnis: Was wird beim Bereinigen geprüft?

Erkläre:

- Welche Bedingungen müssen erfüllt sein, damit ein Datensatz in `bereinigt` aufgenommen wird?
- Warum prüfen wir, ob `ist_zahl(wort)` *nicht* zutrifft?
- Warum wäre `zahl.isdigit()` für die Antwort „7.5“ ungeeignet?
- Was passiert mit Zeilen, in denen die Felder vertauscht sind?

Teste: Sind alle falschen Einträge herausgefiltert, also sind deine Daten jetzt komplett bereinigt?

```python
# Zum Ausprobieren:
print("Vorher:", len(daten), "Datensätze")
print("Nachher:", len(bereinigt), "Datensätze")
```

## 4. Daten analysieren

```python
zaehler = {}
for z in bereinigt:
    antwort = z[1]
    if antwort in zaehler:
        zaehler[antwort] += 1
    else:
        zaehler[antwort] = 1

zahlen = [float(z[2]) for z in bereinigt]
durchschnitt = sum(zahlen) / len(zahlen)

print("Antworten (Wort):", zaehler)
print("Durchschnitt (Zahl):", round(durchschnitt, 1))
```

### Verständnis: Wie greife ich auf Werte zu?

Versuche Folgendes in deinem Programm:

- `print(bereinigt[1])` – Was wird ausgegeben?
- `print(bereinigt[1][2])` – Was steht an dieser Stelle?
- `print(bereinigt[-1][1])` – Was ist das?

Erkläre: Wie funktionieren die Zahlen in den eckigen Klammern? Was bedeuten sie?

```python
print(bereinigt[1])
print(bereinigt[1][2])
print(bereinigt[-1][1])
```

## 5. Neue CSV-Datei speichern

```python
with open('umfrage_bereinigt.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file, delimiter=';')
    writer.writerow(['Nr', 'Antwort_Wort', 'Antwort_Zahl'])
    writer.writerows(bereinigt)
```
