# Geschachtelte Datentypen

In vielen Programmiersprachen lassen sich Datentypen ineinander verschachteln, um komplexe Strukturen abzubilden. Besonders in Python sind geschachtelte Listen und Dictionaries sehr mächtig und weit verbreitet.

## 1. Listen in Listen

Eine Liste kann andere Listen enthalten. Dies erlaubt z. B. die Darstellung von Tabellen oder Matrizen.

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix)
````

## 2. Listen in Dictionaries

Ein Dictionary kann Listen als Werte enthalten. Dies ist praktisch, wenn mehreren Kategorien mehrere Elemente zugeordnet werden sollen.

````python
noten = {"Anna": [2, 1, 3], "Ben": [3, 2, 2]}
print(noten)
````

## 3. Dictionaries in Dictionaries

Auch Dictionaries können verschachtelt werden, z. B. zur Modellierung komplexer Objekte.

````python
personen = {
    "anna": {"alter": 16, "klasse": "10A"},
    "ben": {"alter": 15, "klasse": "9B"}
}
print(personen)
````

## 4. Alles gemixt

Die Kombination aller Strukturen erlaubt es, realitätsnahe Datenmodelle zu erstellen.

````python
schule = {
    "10A": [{"name": "Anna", "noten": [2, 1, 3]}],
    "9B": [{"name": "Ben", "noten": [3, 2, 2]}]
}
print(schule)
````

**Merke:** Geschachtelte Strukturen helfen dabei, komplexe Daten realitätsnah abzubilden. Sie sind Grundlage für die Arbeit mit APIs, Datenbanken und modernen Anwendungen.

## Warum ist das wichtig?

In der realen Softwareentwicklung werden Daten oft in komplexer Form gespeichert, z. B. als JSON. Das Verständnis von geschachtelten Datentypen ist essenziell für:

- Datenanalyse
- Webentwicklung (z. B. REST APIs)
- Arbeit mit Datenbanken
- Machine Learning

````python
# Aufgabe 2
umsätze = [[10.5, 20.0], [5.5, 7.0, 3.0], [12.0]]

gesamtsumme = 0  # Gesamtsumme aller Umsätze

for einkauf in umsätze:  # äußere Schleife: Einkäufe
    tagessumme = 0
    for einzelposten in einkauf:  # innere Schleife: Einzelposten
        tagessumme += einzelposten  # zur Tagessumme hinzufügen
    print(f"Tagesumsatz: {tagessumme} €")
    gesamtsumme += tagessumme  # zur Gesamtsumme hinzufügen

print(f"Gesamtumsatz: {gesamtsumme} €")
````

````python
# Aufgabe 5
# Strukturierte Klassenverwaltung:
# In Python kannst du Dictionaries direkt als Elemente in einer Liste schreiben.
# Du brauchst ihnen keinen Namen zu geben, weil die Liste sie "verwalten" kann

Klassenverwaltung = {
    "Klasse1": [
        {"name": "Anna", "alter": 13, "Noten": [1, 3, 2, 1]},
        {"name": "Max", "alter": 12, "Noten": [1, 1, 1, 1]},
        {"name": "Tom", "alter": 13, "Noten": [4, 3, 3, 2]},
        {"name": "Kira", "alter": 12, "Noten": [2, 3, 4, 3]}
    ],
    "Klasse2": [
        {"name": "Anne", "alter": 13, "Noten": [1, 3, 2, 1]},
        {"name": "Maxi", "alter": 12, "Noten": [1, 1, 1, 1]},
        {"name": "Tim", "alter": 13, "Noten": [4, 3, 3, 2]},
        {"name": "Kita", "alter": 12, "Noten": [2, 3, 4, 3]}
    ]
}

schueler_ges = "Anna"
gefunden = False  # Variable zur Fallunterscheidung

for klasse in Klassenverwaltung:  # Zugriff auf die Klassen
    for schueler in Klassenverwaltung[klasse]:  # Zugriff auf die Schüler einer Klasse
        if schueler["name"] == schueler_ges:  # Abgleich gesuchter Schüler
            durchschnitt = sum(schueler["Noten"]) / len(schueler["Noten"])  # Durchschnitt berechnen
            print(f"Durchschnitt von {schueler['name']}: {durchschnitt}")
            gefunden = True
if gefunden == False:
    print(f"Schüler '{schueler_ges}' nicht gefunden.")
````

````python
# Aufgabe 6
gesuchte_Note = 1
gefunden = False  # Variable zur Fallunterscheidung
Liste_Schueler = []  # Speicherung der Schülernamen

for klasse in Klassenverwaltung:  # iterieren durch die Klassen
    for schueler in Klassenverwaltung[klasse]:  # iterieren durch die Schüler einer Klasse
        for note in schueler["Noten"]:  # iterieren durch die einzelnen Noten der Schüler
            if note == gesuchte_Note:  # Abgleich mit gesuchter Note
                Liste_Schueler.append(schueler["name"])  # Speichern der Schülernamen
                gefunden = True
                break  # Abbruch, sobald einmal die gesuchte Note gefunden wurde (min. eine 1)

print(f"Liste aller Schüler mit mindestens einer Note {gesuchte_Note}: {Liste_Schueler}")

if gefunden == False:
    print(f"Kein Schüler mit Note '{gesuchte_Note}' gefunden.")
````
