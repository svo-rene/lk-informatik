# Programmierparadigmen

In diesem Notebook werden wir verschiedene Programmierparadigmen kennenlernen. Jedes Paradigma wird kurz erklärt und mit Beispielen in verschiedenen Programmiersprachen verdeutlicht.


## 1. Imperatives Paradigma

Das imperative Paradigma beschreibt, wie ein Programm den Zustand eines Computers durch Befehle ändert. Das Programm besteht aus Anweisungen, die nacheinander ausgeführt werden.

```python
# Beispiel: JavaScript
for (let i = 0; i < 5; i++) {
    console.log(i);
}
```

```python
# Beispiel: Java
public class ImperativeExample {
    public static void main(String[] args) {
        for (int i = 0; i < 5; i++) {
            System.out.println(i);
        }
    }
}
```

```python
# Beispiel: Python
for i in range(5):
    print(i)
```

## 2. Deklaratives Paradigma

Im deklarativen Paradigma wird beschrieben, was das Programm tun soll, ohne dass spezifiziert wird, wie es ausgeführt werden soll. SQL ist ein Beispiel für eine deklarative Sprache.

```python
# Beispiel: SQL
SELECT name 
FROM users 
WHERE age = 25;
```

## 3. Funktionales Paradigma

Das funktionale Paradigma basiert auf der Idee der Verwendung von Funktionen als zentrale Bausteine.

```python
# Beispiel: Haskell
fac :: Int -> Int
fac 0 = 1
fac n = n * fac (n - 1)
```

```python
# Beispiel: Python
def fac(n):
    if n == 0: 
        return 1
    else:
        return n * fac(n - 1)

print(fac(5))  # Ausgabe: 120
```

## 4. Logisches Paradigma

Das logische Paradigma basiert auf dem Prinzip der Schlussfolgerung. Prolog ist eine der bekanntesten Sprachen in diesem Paradigma. Logische Programme bestehen aus Fakten und Regeln.

```python
# Beispiel: Prolog
parent(john, mary).
parent(mary, susan).

grandparent(X, Y) :- parent(X, Z), parent(Z, Y).
```

## 5. Objektorientiertes Paradigma

Im objektorientierten Paradigma werden Programme in Objekte unterteilt, die sowohl Daten (Attribute) als auch Funktionen (Methoden) enthalten. Die Objekte interagieren miteinander, um das Verhalten eines Systems abzubilden.

```python
# Beispiel: Python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"

dog = Dog("Buddy")
print(dog.speak())  # Ausgabe: Buddy says Woof!
```
