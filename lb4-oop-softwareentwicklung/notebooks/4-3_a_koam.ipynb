# Grundlagen der OOP mit KOAM
In diesem Notebook lernen wir die Grundlagen der objektorientierten Programmierung (OOP) kennen. Dazu gehören die vier wesentlichen Konzepte der OOP: 
1. **Klasse**
2. **Objekt**
3. **Attribut**
4. **Methode**

### 1. Klasse
Eine Klasse ist eine Vorlage oder ein Bauplan, der beschreibt, wie ein Objekt aufgebaut ist. Klassen legen fest, welche Attribute (Daten) und Methoden (Funktionen) ein Objekt besitzt.

### 2. Objekt
Ein Objekt ist eine Instanz einer Klasse. Es enthält konkrete Werte für die in der Klasse definierten Attribute und kann die in der Klasse definierten Methoden ausführen.

### 3. Attribut
Ein Attribut ist eine Variable, die in einer Klasse definiert ist. Jedes Objekt kann unterschiedliche Werte für seine Attribute haben.

### 4. Methode
Eine Methode ist eine Funktion, die in einer Klasse definiert ist. Sie kann auf die Attribute eines Objekts zugreifen und sie verändern.

```python
# Beispiel einer Klasse 'Auto'
class Auto:
    def __init__(self, marke, baujahr):
        self.marke = marke  # Attribut 'marke'
        self.baujahr = baujahr  # Attribut 'baujahr'
        self.geschwindigkeit = 0  # Attribut 'geschwindigkeit'
        
    def beschleunigen(self, wert):
        '''Methode zum Beschleunigen des Autos'''
        self.geschwindigkeit += wert
        print(f'Das Auto beschleunigt auf {self.geschwindigkeit} km/h.')
        
    def bremsen(self, wert):
        '''Methode zum Bremsen des Autos'''
        self.geschwindigkeit -= wert
        if self.geschwindigkeit < 0:
            self.geschwindigkeit = 0
        print(f'Das Auto bremst auf {self.geschwindigkeit} km/h ab.')
```

### Erstellung eines Objekts
Wir erstellen nun ein Objekt der Klasse `Auto` und führen einige Methoden darauf aus.

```python
# Erstellen eines Objekts
mein_auto = Auto('BMW', 2020)
print(f'Marke: {mein_auto.marke}, Baujahr: {mein_auto.baujahr}')

# Methoden aufrufen
mein_auto.beschleunigen(50)
mein_auto.bremsen(20)
```
